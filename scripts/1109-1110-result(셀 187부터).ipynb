```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

files = [
    "diag/diag_phase0.py",
    "diag/harness.py",
    "diag/phase0_nav.md",
]

for name in files:
    path = root / name
    print(f"\n{'=' * 70}")
    print(f"파일: {name}")
    print(f"크기: {path.stat().st_size:,} bytes")

    lines = path.read_text(encoding="utf-8").splitlines()
    print(f"전체 줄 수: {len(lines)}")

    if name.endswith(".md"):
        for i, line in enumerate(lines[:60], 1):
            print(f"{i:4}: {line}")
    else:
        for i, line in enumerate(lines, 1):
            stripped = line.strip()
            if (
                stripped.startswith(("def ", "class ", "if __name__"))
                or stripped.startswith(("import ", "from "))
            ):
                print(f"{i:4}: {line}")
```

```python
from pathlib import Path
import tarfile

archive = Path(
    "/kaggle/input/datasets/jimin562/capstone1/b1_retrained.tgz"
)
out_root = Path("/kaggle/working/nav_baseline_ckpts/B1")

targets = [
    "c4ma/phr5_d96100_3d_ho37bc_b16_l0.5_e100_gNone_20261005_152953/",
    "c4ma/phr5_d96100_3d_ho38bc_b16_l0.5_e100_gNone_20261004_225155/",
]

out_root.mkdir(parents=True, exist_ok=True)

with tarfile.open(archive, "r:gz") as tar:
    members = [
        m for m in tar.getmembers()
        if any(m.name.startswith(t) for t in targets)
    ]

    for m in members:
        # 지정한 두 폴더의 파일만 추출
        if m.isfile() and m.name.endswith(
            ("best_model.pth", "training_configure.json")
        ):
            tar.extract(m, path=out_root)

print("===== 추출 결과 =====")
for p in sorted(out_root.rglob("*")):
    if p.is_file():
        print(p, f"{p.stat().st_size / 1024**2:.2f} MB")
```

```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

path = root / "diag" / "harness.py"
lines = path.read_text(encoding="utf-8").splitlines()

# 함수 정의 위치 찾기
targets = ["class ModelLoc", "def nav_run", "def make_traj"]

for target in targets:
    matches = [
        i for i, line in enumerate(lines)
        if target in line
    ]

    print(f"\n===== {target} =====")

    if not matches:
        print("정의를 찾지 못했습니다.")
        continue

    for start in matches:
        end = min(start + 85, len(lines))
        for i in range(start, end):
            print(f"{i + 1:4}: {lines[i]}")
```

```python
from pathlib import Path
import ast

root = Path("/kaggle/input/datasets/jimin562/capstone1/capstone1-main/capstone1-main")
path = root / "diag" / "harness.py"
source = path.read_text(encoding="utf-8")
lines = source.splitlines()
tree = ast.parse(source)

targets = {"CorrPred", "CorrLoc", "load_any"}

for node in tree.body:
    if isinstance(node, (ast.ClassDef, ast.FunctionDef)) and node.name in targets:
        print(f"\n===== {node.name} ({node.lineno}~{node.end_lineno}) =====")
        for i in range(node.lineno - 1, node.end_lineno):
            print(f"{i+1:4}: {lines[i]}")
```

```python
from pathlib import Path
import tarfile

archive = Path("/kaggle/input/datasets/jimin562/capstone1/b1_retrained.tgz")
out_dir = Path("/kaggle/working/b1_checkpoints")

targets = [
    "c4ma/phr5_d96100_3d_ho37bc_b16_l0.5_e100_gNone_20261005_152953/best_model.pth",
    "c4ma/phr5_d96100_3d_ho38bc_b16_l0.5_e100_gNone_20261004_225155/best_model.pth",
]

out_dir.mkdir(parents=True, exist_ok=True)

with tarfile.open(archive, "r:gz") as tar:
    for name in targets:
        member = tar.getmember(name)
        source = tar.extractfile(member)
        if source is None:
            raise RuntimeError(f"추출할 수 없음: {name}")

        destination = out_dir / Path(name).parent.name / "best_model.pth"
        destination.parent.mkdir(parents=True, exist_ok=True)
        destination.write_bytes(source.read())

        print(f"추출 완료: {destination}")
        print(f"파일 크기: {destination.stat().st_size:,} bytes")

print("\nB1 체크포인트 추출 완료")
```

```python
from pathlib import Path
import shutil

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

work_root = Path("/kaggle/working/capstone1_nav_seed7")
source_diag = root / "diag"
target_diag = work_root / "diag"

if work_root.exists():
    raise FileExistsError(
        f"{work_root}가 이미 존재해. 기존 작업물을 덮어쓰지 않도록 중단했어."
    )

# 원본 diag 폴더를 작업용 경로로 복사
work_root.mkdir(parents=True)
shutil.copytree(source_diag, target_diag)

harness = target_diag / "harness.py"
source = harness.read_text(encoding="utf-8")

replacements = {
    "nav_traj=96": "nav_traj=32",
    "make_traj(n_traj, 1000, 2026)": "make_traj(n_traj, 1000, 7)",
    "T5='make_traj(n, 1000, 2026)'": "T5='make_traj(n, 1000, 7)'",
}

for old, new in replacements.items():
    count = source.count(old)
    print(f"{old} 발견 횟수: {count}")
    if count != 1:
        raise RuntimeError(f"예상과 다른 횟수라 수정 중단: {old}")
    source = source.replace(old, new)

harness.write_text(source, encoding="utf-8")

print("\n===== 수정된 설정 확인 =====")
for i, line in enumerate(source.splitlines(), 1):
    if "nav_traj=32" in line or "make_traj(n_traj" in line or "T5='make_traj" in line:
        print(f"{i}: {line}")

print("\n작업용 파일:", harness)
print("원본 파일은 수정하지 않았어.")
```

```python
import tarfile
from pathlib import Path

archive = "/kaggle/input/datasets/jimin562/capstone1/b1_retrained.tgz"

restore = {
    "c4ma/phr5_d96100_3d_ho37bc_b16_l0.5_e100_gNone_20261005_152953/training_configure.json":
        "/kaggle/working/b1_checkpoints/phr5_d96100_3d_ho37bc_b16_l0.5_e100_gNone_20261005_152953/training_configure.json",

    "c4ma/phr5_d96100_3d_ho38bc_b16_l0.5_e100_gNone_20261004_225155/training_configure.json":
        "/kaggle/working/b1_checkpoints/phr5_d96100_3d_ho38bc_b16_l0.5_e100_gNone_20261004_225155/training_configure.json",
}

with tarfile.open(archive, "r:gz") as tar:
    for source, destination in restore.items():
        data = tar.extractfile(source).read()
        dest = Path(destination)
        dest.parent.mkdir(parents=True, exist_ok=True)
        dest.write_bytes(data)

        print(f"복원 완료: {dest.name} ({len(data):,} bytes)")

# 복원 결과 확인
for destination in restore.values():
    p = Path(destination)
    print(f"{p.parent.name}: 설정 파일 존재={p.exists()}, 크기={p.stat().st_size:,} bytes")
```

```python
import json
from pathlib import Path

out = Path("/kaggle/working/nav_baseline_seed7")

files = [
    "B1_Singapore_seed7_n32_3d_ho37bc.json",
    "Corr_Singapore_seed7_n32_3d_ho37bc.json",
    "B1_SF_seed7_n32_3d_ho38bc.json",
    "Corr_SF_seed7_n32_3d_ho38bc.json",
]

for name in files:
    path = out / name
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)

    print(f"\n===== {name} =====")
    print("최상위 자료형:", type(data).__name__)

    if isinstance(data, dict):
        print("최상위 키:", list(data.keys()))

        for key, value in data.items():
            if isinstance(value, list):
                print(f"{key}: list, 길이={len(value)}")
                if value:
                    print("첫 항목:", value[0])
                break
            elif isinstance(value, dict):
                print(f"{key}: dict, 하위 키={list(value.keys())[:15]}")
```

**각 JSON의 T5 데이터 구조와 궤적 개수 확인**

```python
import json
from pathlib import Path

out = Path("/kaggle/working/nav_baseline_seed7")

files = [
    "B1_Singapore_seed7_n32_3d_ho37bc.json",
    "Corr_Singapore_seed7_n32_3d_ho37bc.json",
    "B1_SF_seed7_n32_3d_ho38bc.json",
    "Corr_SF_seed7_n32_3d_ho38bc.json",
]

for name in files:
    with open(out / name, "r", encoding="utf-8") as f:
        data = json.load(f)

    print(f"\n===== {name} =====")

    samples = data["samples"]
    t5 = samples["T5"]

    print("T5 자료형:", type(t5).__name__)

    if isinstance(t5, dict):
        print("T5 키:", list(t5.keys()))

        for key, value in t5.items():
            print(f"  {key}: {type(value).__name__}", end="")
            if isinstance(value, (list, dict)):
                print(f", 길이={len(value)}")
                if len(value) > 0:
                    print("    첫 항목:", str(value[0] if isinstance(value, list) else next(iter(value.items())))[:300])
            else:
                print(":", str(value)[:200])

    elif isinstance(t5, list):
        print("T5 길이:", len(t5))
        if t5:
            print("첫 항목:", str(t5[0])[:500])

    nav = data["T5_nav"]
    print("T5_nav 키:", list(nav.keys()))

    for key, value in nav.items():
        print(f"  {key}: {type(value).__name__}")
        if isinstance(value, dict):
            print("    하위 키:", list(value.keys()))
```

**각 실행 결과의 steps CSV를 비교**

```python
import pandas as pd
from pathlib import Path

out = Path("/kaggle/working/nav_baseline_seed7/steps")

pairs = [
    ("Singapore",
     "B1_Singapore_seed7_n32_3d_ho37bc.csv",
     "Corr_Singapore_seed7_n32_3d_ho37bc.csv"),
    ("SF",
     "B1_SF_seed7_n32_3d_ho38bc.csv",
     "Corr_SF_seed7_n32_3d_ho38bc.csv"),
]

for city, b1_file, corr_file in pairs:
    b1 = pd.read_csv(out / b1_file)
    corr = pd.read_csv(out / corr_file)

    print(f"\n{'='*15} {city} {'='*15}")

    for name, df in [("B1", b1), ("CorrMatchNet", corr)]:
        first = df.sort_values("step").groupby("traj").first()
        last = df.sort_values("step").groupby("traj").last()

        print(f"\n[{name}]")
        print("고유 궤적 수:", df["traj"].nunique())
        print("reason 분포:", df.groupby("traj")["reason"].last().value_counts().to_dict())
        print("궤적별 평균 기록 행 수:", round(len(df) / df["traj"].nunique(), 2))

        print("첫 행의 시작 좌표 (traj 0~2):")
        print(first.loc[[0, 1, 2], ["real_x", "real_y", "block_x", "block_y"]])

        print("궤적별 마지막 기록 (traj 0~2):")
        print(last.loc[[0, 1, 2], ["step", "real_x", "real_y", "success", "reason"]])

    b1_first = b1.sort_values("step").groupby("traj").first()
    corr_first = corr.sort_values("step").groupby("traj").first()

    common = b1_first.index.intersection(corr_first.index)

    same_start = (
        b1_first.loc[common, ["real_x", "real_y", "block_x", "block_y"]]
        .round(6)
        .equals(
            corr_first.loc[common, ["real_x", "real_y", "block_x", "block_y"]]
            .round(6)
        )
    )

    print("\n시작 좌표 전체 일치:", same_start)
    print("공통 궤적 수:", len(common))
```

**(3) 궤적별 쌍체 비교 및 통계 검정**

```python
import pandas as pd
from pathlib import Path
from scipy.stats import binomtest

out = Path("/kaggle/working/nav_baseline_seed7/steps")

pairs = [
    ("Singapore",
     "B1_Singapore_seed7_n32_3d_ho37bc.csv",
     "Corr_Singapore_seed7_n32_3d_ho37bc.csv"),
    ("SF",
     "B1_SF_seed7_n32_3d_ho38bc.csv",
     "Corr_SF_seed7_n32_3d_ho38bc.csv"),
]

for city, b1_file, corr_file in pairs:
    b1 = pd.read_csv(out / b1_file)
    corr = pd.read_csv(out / corr_file)

    b1_result = b1.groupby("traj")["success"].max().rename("B1")
    corr_result = corr.groupby("traj")["success"].max().rename("CorrMatchNet")

    paired = pd.concat([b1_result, corr_result], axis=1).dropna()
    paired["B1"] = paired["B1"].astype(int)
    paired["CorrMatchNet"] = paired["CorrMatchNet"].astype(int)

    both_success = ((paired.B1 == 1) & (paired.CorrMatchNet == 1)).sum()
    b1_only = ((paired.B1 == 1) & (paired.CorrMatchNet == 0)).sum()
    corr_only = ((paired.B1 == 0) & (paired.CorrMatchNet == 1)).sum()
    both_fail = ((paired.B1 == 0) & (paired.CorrMatchNet == 0)).sum()

    discordant = b1_only + corr_only
    p_value = (
        binomtest(corr_only, discordant, 0.5).pvalue
        if discordant > 0 else 1.0
    )

    print(f"\n===== {city}: 궤적별 쌍체 비교 =====")
    print("비교 궤적 수:", len(paired))
    print("둘 다 성공:", both_success)
    print("B1만 성공:", b1_only)
    print("CorrMatchNet만 성공:", corr_only)
    print("둘 다 실패:", both_fail)
    print(f"정확 이항검정 p-value: {p_value:.8g}")

    print("\n궤적별 결과:")
    display(paired)
```

**결과 정리 1차**

```python
from pathlib import Path
import shutil
import pandas as pd

src = Path("/kaggle/working/nav_baseline_seed7")
dst = Path("/kaggle/working/capstone1_nav_seed7/experiments/2026-11-13_nav_baselines")

# 결과 폴더 생성
for model in ["B1", "CorrMatchNet"]:
    (dst / model).mkdir(parents=True, exist_ok=True)

# 원본 결과 파일 복사
mapping = {
    "B1": {
        "Singapore": "B1_Singapore_seed7_n32_3d_ho37bc",
        "SF": "B1_SF_seed7_n32_3d_ho38bc",
    },
    "CorrMatchNet": {
        "Singapore": "Corr_Singapore_seed7_n32_3d_ho37bc",
        "SF": "Corr_SF_seed7_n32_3d_ho38bc",
    },
}

for model, cities in mapping.items():
    for city, stem in cities.items():
        shutil.copy2(
            src / "steps" / f"{stem}.csv",
            dst / model / f"{city}_steps.csv"
        )
        shutil.copy2(
            src / f"{stem}.json",
            dst / model / f"{city}_result.json"
        )

# 모델별 요약표
rows = [
    ["B1", "Singapore", 7, 32, 18.8, "8.9–35.3", 35.5, 26, 6],
    ["CorrMatchNet", "Singapore", 7, 32, 100.0, "89.3–100.0", 50.7, 0, 32],
    ["B1", "SF", 7, 32, 40.6, "25.5–57.7", 29.0, 19, 13],
    ["CorrMatchNet", "SF", 7, 32, 100.0, "89.3–100.0", 52.4, 0, 32],
]

columns = [
    "model", "city", "seed", "trajectories",
    "success_pct", "ci95_pct", "runtime_sec",
    "out_of_map", "arrived_ok"
]

summary = pd.DataFrame(rows, columns=columns)

for model in ["B1", "CorrMatchNet"]:
    summary[summary["model"] == model].to_csv(
        dst / model / "summary.csv",
        index=False,
        encoding="utf-8-sig"
    )

# 전체 비교표
summary.to_csv(
    dst / "summary.csv",
    index=False,
    encoding="utf-8-sig"
)

print("저장 위치:", dst)
print("\n===== 전체 요약 =====")
display(summary)

print("\n===== 생성 파일 =====")
for p in sorted(dst.rglob("*")):
    if p.is_file():
        print(p.relative_to(dst), f"({p.stat().st_size:,} bytes)")
```

```python
from pathlib import Path
import pandas as pd

dst = Path("/kaggle/working/capstone1_nav_seed7/experiments/2026-11-13_nav_baselines")
steps_dir = dst

# 1. 궤적별 쌍체 비교 CSV 생성
paired_tables = []

for city in ["Singapore", "SF"]:
    b1 = pd.read_csv(dst / "B1" / f"{city}_steps.csv")
    corr = pd.read_csv(dst / "CorrMatchNet" / f"{city}_steps.csv")

    b1_result = b1.groupby("traj")["success"].max().rename("B1")
    corr_result = corr.groupby("traj")["success"].max().rename("CorrMatchNet")

    paired = pd.concat([b1_result, corr_result], axis=1).reset_index()
    paired.insert(0, "city", city)
    paired_tables.append(paired)

paired_all = pd.concat(paired_tables, ignore_index=True)
paired_all.to_csv(dst / "paired_comparison.csv", index=False, encoding="utf-8-sig")

# 2. 비교 보고서 Markdown 생성
report = r"""# B1 vs CorrMatchNet 항법 기준선 비교

## 1. 실험 목적

동일한 초기 항법 궤적 조건에서 B1과 CorrMatchNet의 폐루프 항법 성공률을 비교한다.
평가는 Singapore와 San Francisco(SF) 두 도시에서 수행하였다.

## 2. 실험 조건

| 항목 | 설정 |
|---|---|
| 평가 항목 | T5 navigation |
| 모델 | B1, CorrMatchNet |
| 도시 | Singapore (`37bc`), SF (`38bc`) |
| Seed | 7 |
| 궤적 수 | 도시별 32개 |
| View 설정 | `3d` |
| 데이터 크기 설정 | `full` |
| 이동 간격 | 15 m |
| 도착 판정 거리 | 20 m |
| 초기 heading | 90° |
| 장치 | Kaggle GPU 환경 |

**입력 및 실험 범위 주의:** 이 실험은 실제 3D UAV 영상 입력을 이용한 항법 실험이 아니다.
`3d` 설정은 3D-view 체크포인트 및 메타데이터 선택에 사용되었으며,
T5 항법 과정은 2D 지도 crop 기반으로 수행되었다.
따라서 결과는 '3D-view 체크포인트/메타데이터를 선택한 2D 지도 기반 폐루프 항법 실험'으로 해석해야 한다.

## 3. 전체 결과

| 도시 | 모델 | 성공 | 성공률 | 95% CI | 실행 시간(초) |
|---|---|---:|---:|---:|---:|
| Singapore | B1 | 6/32 | 18.8% | 8.9–35.3% | 35.5 |
| Singapore | CorrMatchNet | 32/32 | 100.0% | 89.3–100.0% | 50.7 |
| SF | B1 | 13/32 | 40.6% | 25.5–57.7% | 29.0 |
| SF | CorrMatchNet | 32/32 | 100.0% | 89.3–100.0% | 52.4 |

### 실패 원인 분포

| 도시 | 모델 | `arrived_ok` | `out_of_map` |
|---|---|---:|---:|
| Singapore | B1 | 6 | 26 |
| Singapore | CorrMatchNet | 32 | 0 |
| SF | B1 | 13 | 19 |
| SF | CorrMatchNet | 32 | 0 |

## 4. 동일 궤적 기준 쌍체 비교

두 모델은 도시별 동일한 seed와 32개 궤적을 사용했다.
단계 기록 CSV에서 궤적별 첫 시작 좌표(`real_x`, `real_y`)와 시작 블록(`block_x`, `block_y`)을 비교했으며,
두 모델의 시작 조건이 모두 일치함을 확인했다.

| 도시 | 둘 다 성공 | B1만 성공 | CorrMatchNet만 성공 | 둘 다 실패 |
|---|---:|---:|---:|---:|
| Singapore | 6 | 0 | 26 | 0 |
| SF | 13 | 0 | 19 | 0 |

불일치 쌍에 대한 정확 이항검정 결과:

| 도시 | 정확 이항검정 p-value |
|---|---:|
| Singapore | 2.9802322 × 10⁻⁸ |
| SF | 3.8146973 × 10⁻⁶ |

두 도시 모두 p-value가 0.05보다 작다.
따라서 이번 seed 7의 32개 궤적에서는 CorrMatchNet의 성공 여부가 B1보다 유의하게 높은 방향으로 나타났다.

## 5. 실행 시간

| 도시 | B1 (초) | CorrMatchNet (초) |
|---|---:|---:|
| Singapore | 35.5 | 50.7 |
| SF | 29.0 | 52.4 |

CorrMatchNet은 두 도시 모두 B1보다 전체 실행 시간이 길었다.
다만 실행 시간은 해당 실험 실행에서 기록된 값이며, 다른 GPU나 실행 환경에 그대로 일반화할 수 없다.

## 6. 해석 및 한계

1. 두 도시 모두 CorrMatchNet은 32개 궤적에서 100% 성공했고, B1은 Singapore 18.8%, SF 40.6%를 기록했다.
2. 동일 궤적 쌍체 비교에서 B1만 성공한 궤적은 없었으며, CorrMatchNet만 성공한 궤적은 Singapore 26개, SF 19개였다.
3. 두 도시의 쌍체 비교 검정은 통계적으로 유의했다.
4. 다만 평가는 도시별 seed 7, 32개 궤적에 한정된다. 여러 seed와 더 많은 궤적에서의 재현성은 아직 검증하지 않았다.
5. 본 실험은 실제 3D UAV 영상 입력을 사용하지 않은 2D 지도 기반 항법 실험이다. 따라서 실제 3D 영상 기반 항법 성능으로 확대 해석하면 안 된다.
6. 시작 조건은 일치하지만, 모델의 위치 추정 결과가 이후 이동에 영향을 주므로 전체 이동 경로가 동일한 것은 아니다.

## 7. 결론

Seed 7의 동일 초기 궤적 32개를 대상으로 한 T5 평가에서 CorrMatchNet은 Singapore와 SF 모두 100% 성공률을 기록해 B1보다 높은 항법 성공률을 보였다.
쌍체 비교에서도 CorrMatchNet만 성공한 궤적이 두 도시에서 각각 26개와 19개였고, 정확 이항검정 결과는 모두 통계적으로 유의했다.
다만 결과의 적용 범위는 현재 평가한 2개 도시, seed 7, 도시별 32개 궤적 및 2D 지도 기반 항법 설정으로 제한한다.
"""

(dst / "comparison.md").write_text(report, encoding="utf-8")

print("저장 완료")
print(" -", dst / "comparison.md")
print(" -", dst / "paired_comparison.csv")

print("\n===== 쌍체 비교 CSV 미리보기 =====")
display(paired_all.head(10))

print("\n===== 생성 파일 확인 =====")
for name in ["comparison.md", "paired_comparison.csv"]:
    p = dst / name
    print(name, "| 존재:", p.exists(), "| 크기:", p.stat().st_size, "bytes")
```

```python
import json
from pathlib import Path
from collections import Counter

root = Path(DATA_ROOT)
uav_dir = next((root / "citya").glob("uav_*"))
json_files = sorted(uav_dir.rglob("*.json"))

key_counts = Counter()

for path in json_files[:100]:
    with open(path, "r", encoding="utf-8") as f:
        data = json.load(f)
    key_counts.update(data.keys())

print("===== JSON 키 목록 및 출현 횟수 =====")
for key, count in key_counts.most_common():
    print(f"{key:20s}: {count}/100")
```

```python
from pathlib import Path
import os

search_roots = [
    Path("/kaggle/input"),
    Path("/kaggle/working"),
]

weight_exts = {".pth", ".pt", ".ckpt", ".safetensors"}

print("===== 모델 체크포인트 검색 =====")

found = []

for root in search_roots:
    if not root.exists():
        continue

    for current, dirs, files in os.walk(root):
        # 이미지 데이터 폴더는 검색에서 제외
        dirs[:] = [
            d for d in dirs
            if not d.startswith("sat_")
            and not d.startswith("uav_")
        ]

        for filename in files:
            path = Path(current) / filename

            if path.suffix.lower() in weight_exts:
                size_gb = path.stat().st_size / (1024 ** 3)
                found.append((path, size_gb))

for path, size_gb in sorted(found, key=lambda x: str(x[0])):
    print(f"{size_gb:.3f} GB | {path}")

print(f"\n체크포인트 파일 수: {len(found)}")
```

```python
import torch
from pathlib import Path

from diag.diag_modules import load_model, DEV
from cvphr.models.corrmatch import CorrMatchNet

# 체크포인트 경로
b1_dir = Path(
    "/kaggle/working/b1_retrained/"
    "c4ma/phr5_d96100_3d_ho38bc_b16_l0.5_e100_gNone_20261004_225155"
)

corr_dir = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main/results/corr/"
    "p2bv2_3d_ho38bc_20261008_131032"
)

print("GPU:", torch.cuda.get_device_name(0))
print("Device:", DEV)

# B1 로딩
print("\n===== B1 로딩 =====")
b1_model = load_model(str(b1_dir))
print("B1:", type(b1_model).__name__)
print("B1 eval:", not b1_model.training)

# CorrMatchNet 로딩
print("\n===== CorrMatchNet 로딩 =====")
corr_model = CorrMatchNet.load(
    str(corr_dir),
    device=DEV,
    decode_override=None
)
print("Corr:", type(corr_model).__name__)
print("Corr eval:", not corr_model.training)
print("Corr decode:", corr_model.decode_cfg)
print("Corr confidence temperature:", corr_model.conf_temp)

print("\n===== 완료 =====")
print("두 모델 로딩 확인")
```

```python
import time
import numpy as np
import pandas as pd
from pathlib import Path
from collections import Counter
from diag.diag_phase0 import nav_run

# 결과 저장 경로
out_dir = Path("/kaggle/working/nav_b1_corr_32_seed7")
out_dir.mkdir(parents=True, exist_ok=True)

print("===== B1 3D 항법 시작 =====")
print("모델:", type(b1_model).__name__)
print("지도:", sf_img.shape)
print("궤적 수:", len(starts))
print("목표 거리: 1000m")
print("Seed: 7")

t0 = time.perf_counter()

# B1 항법 실행
b1_success, b1_reason, b1_errors = nav_run(
    b1_loc,
    sf_img,
    starts.copy(),
    goals.copy(),
    decode_px=128,
    oracle_block=False,
    step_m=15.0,
    arrive_m=20.0,
    n_block=15,
)

elapsed = time.perf_counter() - t0

# 결과 정리
b1_success = np.asarray(b1_success, dtype=bool)
b1_reason = np.asarray(b1_reason, dtype=str)
b1_errors = np.asarray(b1_errors, dtype=float)

print("\n===== B1 결과 =====")
print("성공:", int(b1_success.sum()), "/", len(b1_success))
print("성공률:", round(100 * b1_success.mean(), 2), "%")
print("실패:", int((~b1_success).sum()))
print("실패 사유:", Counter(b1_reason[~b1_success]))
print("평균 실행 시간:", round(elapsed, 2), "초")
print("궤적당 시간:", round(elapsed / len(starts), 3), "초")

# CSV 저장
df_b1 = pd.DataFrame({
    "trajectory_id": np.arange(len(b1_success)),
    "success": b1_success,
    "reason": b1_reason,
    "median_step_error": b1_errors,
})

csv_path = out_dir / "B1_3d_ho38bc_32_seed7.csv"
df_b1.to_csv(csv_path, index=False)

print("\n저장 완료:", csv_path)
print(df_b1.head(10).to_string(index=False))
```

**궤적별 paired 비교 및 McNemar 검정**

```python
import pandas as pd
import numpy as np
from scipy.stats import binomtest
from pathlib import Path

out_dir = Path("/kaggle/working/nav_b1_corr_32_seed7")

b1 = pd.read_csv(out_dir / "B1_3d_ho38bc_32_seed7.csv")
corr = pd.read_csv(out_dir / "CorrMatchNet_3d_ho38bc_32_seed7.csv")

# 궤적 ID 일치 확인
assert np.array_equal(b1["trajectory_id"], corr["trajectory_id"])
assert len(b1) == len(corr) == 32

paired = pd.DataFrame({
    "trajectory_id": b1["trajectory_id"],
    "B1_success": b1["success"].astype(bool),
    "Corr_success": corr["success"].astype(bool),
})

# 대응 결과 2x2
both_success = (
    paired["B1_success"] & paired["Corr_success"]
).sum()

b1_only = (
    paired["B1_success"] & ~paired["Corr_success"]
).sum()

corr_only = (
    ~paired["B1_success"] & paired["Corr_success"]
).sum()

both_fail = (
    ~paired["B1_success"] & ~paired["Corr_success"]
).sum()

print("===== Paired 성공 결과 =====")
print("둘 다 성공:", both_success)
print("B1만 성공:", b1_only)
print("CorrMatchNet만 성공:", corr_only)
print("둘 다 실패:", both_fail)

# Exact McNemar test (discordant pairs에 대한 exact binomial test)
n_discordant = b1_only + corr_only

if n_discordant:
    p_value = binomtest(
        min(b1_only, corr_only),
        n=n_discordant,
        p=0.5,
        alternative="two-sided"
    ).pvalue
else:
    p_value = 1.0

print("\n===== Exact McNemar 검정 =====")
print("Discordant pairs:", n_discordant)
print("p-value:", p_value)

print("\n===== 성공률 차이 =====")
print(
    f"CorrMatchNet - B1: "
    f"{100 * (corr['success'].mean() - b1['success'].mean()):.2f}%p"
)

# 궤적별 결과 저장
paired.to_csv(out_dir / "paired_success_32_seed7.csv", index=False)
print("\nPaired 결과 저장 완료")
```

```python
from pathlib import Path

repo = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

targets = {
    "B1": repo / "diag/diag_phase0.py",
    "CorrMatchNet": repo / "diag/harness.py",
}

keywords = {
    "B1": ["class ModelLoc", "def __call__", "def nav_run"],
    "CorrMatchNet": ["class CorrLoc", "def __call__", "def predict"],
}

for model, path in targets.items():
    print("\n" + "=" * 65)
    print(model, "|", path.exists())
    print("=" * 65)

    if not path.exists():
        continue

    lines = path.read_text(encoding="utf-8").splitlines()

    for keyword in keywords[model]:
        matches = [
            i for i, line in enumerate(lines)
            if keyword in line
        ]

        for idx in matches:
            start = max(0, idx - 3)
            end = min(len(lines), idx + 35)

            print(f"\n--- {keyword} (line {idx + 1}) ---")

            for j in range(start, end):
                print(f"{j + 1:4d}: {lines[j]}")
```

```python
from pathlib import Path

repo = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

files = [
    repo / "diag/diag_phase0.py",
    repo / "diag/diag_gluf.py",
]

for path in files:
    print("\n" + "=" * 70)
    print(path.name)
    print("=" * 70)

    lines = path.read_text(encoding="utf-8").splitlines()

    keywords = [
        "class TileCache",
        "def block(",
        "def load_model",
        "def enc(",
        "def head(",
    ]

    found = set()

    for i, line in enumerate(lines):
        if any(k in line for k in keywords):
            key = next(k for k in keywords if k in line)
            if key in found:
                continue
            found.add(key)

            print(f"\n--- {key} | line {i+1} ---")
            for j in range(max(0, i-2), min(len(lines), i+24)):
                print(f"{j+1:4d}: {lines[j]}")
```

```python
from pathlib import Path

roots = [
    Path("/kaggle/input/datasets/jimin562/capstone1/"
         "capstone1-main/capstone1-main"),
    Path("/kaggle/input/datasets/jimin562/"
         "bearing-uav-90k-city-archives")
]

keywords = ["base_00", "base_10", "p1_path", "target_patch_3d"]

for root in roots:
    print("\n===== 검색 위치:", root, "=====")

    if not root.exists():
        print("경로 없음")
        continue

    count = 0
    for path in root.rglob("*.py"):
        try:
            lines = path.read_text(
                encoding="utf-8", errors="ignore"
            ).splitlines()
        except Exception:
            continue

        hits = [
            (i + 1, line.strip())
            for i, line in enumerate(lines)
            if any(k in line for k in keywords)
        ]

        if hits:
            print("\n파일:", path)
            for line_no, line in hits[:12]:
                print(f"  {line_no}: {line[:250]}")
            count += 1

        if count >= 15:
            print("검색 결과 15개 파일에서 중단")
            break
```

```python
import pandas as pd
import numpy as np
import re
from pathlib import Path

path = Path(
    "/kaggle/input/datasets/jimin562/bearing-uav-90k-city-archives/"
    "Bearing_UAV_90K/c4m_254k_96bc_b15_s100_v3d/metadata/metadata.csv"
)

df = pd.read_csv(path)

# 파일명에서 표본 번호 추출
df["sample_no"] = df["target_path"].str.extract(
    r"_s_(\d+)_v3d\.jpg"
).astype(float)

print("번호 추출 성공:", df["sample_no"].notna().sum(), "/", len(df))

# 첫 번째 블록 선택
block = df[
    (df["block_x"] == 0) &
    (df["block_y"] == 0)
].copy()

block = block.sort_values("sample_no").reset_index(drop=True)

# 번호 순서에 따른 위치 변화
block["dx"] = block["x_norm"].diff()
block["dy"] = block["y_norm"].diff()

# 1단위 = 32m 기준
block["distance_m"] = np.sqrt(
    block["dx"]**2 + block["dy"]**2
) * 32

print("\n===== 첫 번째 블록: 번호순 20개 =====")
print(
    block[
        ["sample_no", "x_norm", "y_norm", "theta", "distance_m"]
    ].head(20).to_string(index=False)
)

d = block["distance_m"].dropna()

print("\n===== 번호가 연속된 표본의 위치 변화 =====")
print("표본 수:", len(block))
print("평균 이동 거리:", round(d.mean(), 2), "m")
print("중앙 이동 거리:", round(d.median(), 2), "m")
print("최소 이동 거리:", round(d.min(), 2), "m")
print("최대 이동 거리:", round(d.max(), 2), "m")
print("15m ± 5m 범위:", int(d.between(10, 20).sum()), "/", len(d))

# 번호 간격이 실제로 1인 쌍만 따로 계산
consecutive = block["sample_no"].diff() == 1
dc = block.loc[consecutive, "distance_m"]

print("\n번호가 정확히 1씩 증가하는 쌍:", len(dc))
if len(dc):
    print("평균 거리:", round(dc.mean(), 2), "m")
    print("중앙 거리:", round(dc.median(), 2), "m")
```

```python
from pathlib import Path
import pandas as pd

root = Path(
    "/kaggle/input/datasets/jimin562/"
    "bearing-uav-90k-city-archives/Bearing_UAV_90K"
)

keywords = [
    "time", "timestamp", "traj", "sequence",
    "frame", "flight", "route", "path", "episode"
]

print("===== 전체 CSV 메타데이터 조사 =====")

csv_files = sorted(root.rglob("*.csv"))
print("CSV 파일 수:", len(csv_files))

for path in csv_files:
    try:
        df = pd.read_csv(path, nrows=5)
        cols = df.columns.tolist()

        related = [
            c for c in cols
            if any(k in c.lower() for k in keywords)
        ]

        print("\n파일:", path.relative_to(root))
        print("열:", cols)

        if related:
            print("경로/시간 관련 후보 열:", related)
            print(df[related].head(5).to_string(index=False))

    except Exception as e:
        print("\n읽기 실패:", path.relative_to(root), str(e))
```

```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

keywords = [
    "phr5",
    "b1",
    "load_state_dict",
    "best_model.pth",
    "model_class",
]

print("===== B1 관련 Python 코드 위치 =====")

for path in root.rglob("*.py"):
    try:
        lines = path.read_text(
            encoding="utf-8",
            errors="ignore"
        ).splitlines()
    except Exception:
        continue

    hits = []
    for i, line in enumerate(lines, 1):
        if any(key in line.lower() for key in keywords):
            hits.append((i, line.strip()))

    if hits:
        print(f"\n--- {path.relative_to(root)} ---")
        for i, line in hits[:15]:
            print(f"{i:4d}: {line[:200]}")
```

```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

targets = {
    "diag/diag_modules.py": [(1, 115)],
    "cvphr/models/posaglreg/models.py": [(430, 530), (700, 810), (890, 940)],
    "diag/harness.py": [(330, 390)],
}

for filename, ranges in targets.items():
    path = root / filename
    lines = path.read_text(encoding="utf-8", errors="ignore").splitlines()

    for start, end in ranges:
        print(f"\n{'=' * 18} {filename}:{start}-{end} {'=' * 18}")

        for i in range(start, min(end, len(lines)) + 1):
            print(f"{i:4d}: {lines[i - 1]}")
```

```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

targets = [
    root / "cvphr/models/posaglreg/models.py",
    root / "diag/harness.py",
]

keywords = [
    "class PARCASGM_v5a",
    "def forward",
    "def predict",
    "pos_regressor",
    "dir_regressor",
    "if a.method",
    "nav_run",
]

for path in targets:
    lines = path.read_text(
        encoding="utf-8",
        errors="ignore"
    ).splitlines()

    print(f"\n{'=' * 20} {path.name} {'=' * 20}")

    hit_lines = []
    for i, line in enumerate(lines):
        if any(key in line for key in keywords):
            hit_lines.append(i)

    shown = set()
    for idx in hit_lines:
        start = max(0, idx - 5)
        end = min(len(lines), idx + 16)

        if any(j in shown for j in range(start, end)):
            continue

        print(f"\n--- lines {start + 1}-{end} ---")
        for j in range(start, end):
            print(f"{j + 1:4d}: {lines[j]}")
            shown.add(j)
```

```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

targets = {
    "diag/diag_phase0.py": ["class ModelLoc", "def nav_run"],
    "diag/harness.py": ["class ModelPred", "def load_any"],
    "cvphr/models/posaglreg/models.py": ["def _regress", "def forward_tile_feats"],
}

for filename, keywords in targets.items():
    path = root / filename
    lines = path.read_text(
        encoding="utf-8",
        errors="ignore"
    ).splitlines()

    print(f"\n{'=' * 20} {filename} {'=' * 20}")

    for keyword in keywords:
        hits = [
            i for i, line in enumerate(lines)
            if keyword in line
        ]

        for idx in hits:
            start = max(0, idx - 5)
            end = min(len(lines), idx + 45)

            print(f"\n--- {keyword}: lines {start + 1}-{end} ---")
            for j in range(start, end):
                print(f"{j + 1:4d}: {lines[j]}")
```

```python
from pathlib import Path
import json

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

# 1. 기존 B1 지도 타일 캐시
for filename in ["diag/diag_phase0.py", "diag/diag_nav.py"]:
    path = root / filename
    if not path.exists():
        continue

    lines = path.read_text(
        encoding="utf-8",
        errors="ignore"
    ).splitlines()

    for i, line in enumerate(lines):
        if "class TileCache" in line:
            print(f"\n===== {filename}: TileCache =====")
            for j in range(i, min(i + 65, len(lines))):
                print(f"{j+1:4d}: {lines[j]}")

# 2. imatch1의 3D 렌더 처리 코드
imatch1 = Path(
    "/kaggle/input/datasets/jimin562/"
    "c5-trone-imatch-evaluation/imatch1/nav.py"
)

if imatch1.exists():
    lines = imatch1.read_text(
        encoding="utf-8",
        errors="ignore"
    ).splitlines()

    for keyword in ["class Nearest3D", "class MatchLoc", "class Crop2D"]:
        for i, line in enumerate(lines):
            if keyword in line:
                print(f"\n===== imatch1/nav.py: {keyword} =====")
                for j in range(i, min(i + 55, len(lines))):
                    print(f"{j+1:4d}: {lines[j]}")

# 3. B1의 3D 체크포인트 설정 파일 찾기
print("\n===== B1 3D 체크포인트 설정 =====")

for path in (root / "results/c4ma").glob("*3d_ho*_b16_*"):
    config_path = path / "training_configure.json"

    if not config_path.exists():
        continue

    try:
        config = json.loads(
            config_path.read_text(encoding="utf-8")
        )
        print(f"\n체크포인트: {path.name}")
        print("model_class:", config.get("model_class"))
        print("model_kwargs:", config.get("model_kwargs"))
    except Exception as e:
        print(path.name, "읽기 오류:", e)
```

```python
from pathlib import Path
import tarfile
import json
import cv2
import pandas as pd

# B1 압축파일에서 3D 설정 파일 확인
archive = Path(
    "/kaggle/input/datasets/jimin562/capstone1/b1_retrained.tgz"
)

print("===== B1 체크포인트 설정 =====")

with tarfile.open(archive, "r:gz") as tar:
    members = [
        m for m in tar.getmembers()
        if "3d_ho" in m.name
        and m.name.endswith("training_configure.json")
    ]

    for m in members:
        print("\n파일:", m.name)
        f = tar.extractfile(m)
        config = json.load(f)

        print("model_class:", config.get("model_class"))
        print("model_kwargs:", config.get("model_kwargs"))

# 3D 렌더 이미지 크기 확인
print("\n===== 3D 렌더 이미지 =====")

capstone1 = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

meta = capstone1 / (
    "Bearing_UAV_90K/c4m_254k_96bc_b15_s100_v3d/"
    "metadata/metadata.csv"
)

print("metadata 존재:", meta.exists())

if meta.exists():
    df = pd.read_csv(meta)
    print("전체 렌더 수:", len(df))
    print("컬럼:", df.columns.tolist())

    for _, row in df[df["target_path"].str.contains(
        r"_3d_ho(37|38)bc_", regex=True
    )].head(4).iterrows():

        path = capstone1 / row["target_path"]
        image = cv2.imread(str(path))

        print(
            row["target_path"],
            "| 이미지:", None if image is None else image.shape,
            "| x_norm:", row["x_norm"],
            "| y_norm:", row["y_norm"]
        )
```

```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/maps_fixed/"
    "Bearing_UAV_90K"
)

print("===== Bearing_UAV_90K 구조 =====")

def show_tree(path, depth=0, max_depth=3):
    if depth > max_depth:
        return

    try:
        items = sorted(path.iterdir())
    except Exception as e:
        print("읽기 오류:", e)
        return

    for p in items[:30]:
        print(
            "  " * depth,
            ("[DIR] " if p.is_dir() else "[FILE] "),
            p.name
        )
        if p.is_dir() and depth < max_depth:
            show_tree(p, depth + 1, max_depth)

show_tree(root)

print("\n===== 주요 파일 검색 =====")

for pattern in [
    "metadata.csv",
    "*v3d*",
    "*.jpg",
    "*.png",
    "*.tif",
]:
    found = list(root.rglob(pattern))
    print(f"\n{pattern}: {len(found)}개")
    for p in found[:5]:
        print(" ", p.relative_to(root))
```

```python
from pathlib import Path

root = Path("kaggle/input/datasets/jimin562/c5-trone-imatch-evaluation")

for filename in ["imatch1/nav.py", "imatch1/__init__.py"]:
    p = root / filename
    print(f"\n{'='*20} {filename} {'='*20}")

    if not p.exists():
        print("파일 없음:", p)
        continue

    lines = p.read_text(encoding="utf-8").splitlines()

    if filename.endswith("nav.py"):
        # 항법 실행과 입력 프레임을 만드는 부분
        keys = [
            "class ODO",
            "ODO =",
            "class Nearest3D",
            "class MatchLoc",
            "def nav_run",
        ]
        for key in keys:
            hits = [i for i, line in enumerate(lines) if key in line]
            for start in hits:
                end = min(start + 45, len(lines))
                print(f"\n--- {key}, line {start+1} ---")
                for i in range(start, end):
                    print(f"{i+1:4}: {lines[i]}")
    else:
        print("\n".join(lines))
```

```python
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

targets = [
    root / "cvphr/models/corrmatch.py",
    root / "diag/diag_phase0.py",
    root / "diag/diag_nav.py",
    root / "diag/harness.py",
]

keywords = (
    "class ", "def forward", "def predict",
    "def tile_features", "def forward_tile_feats",
    "def north_up", "PARCASGM_v5a",
    "ModelLoc", "CorrLoc", "checkpoint",
)

for p in targets:
    print(f"\n{'='*25} {p.name} {'='*25}")
    lines = p.read_text(encoding="utf-8").splitlines()

    matched = [
        i for i, line in enumerate(lines)
        if any(k in line for k in keywords)
    ]

    shown = set()
    for i in matched:
        start, end = max(0, i - 3), min(len(lines), i + 9)
        if any(j in shown for j in range(start, end)):
            continue
        print(f"\n--- {start+1}~{end}행 ---")
        for j in range(start, end):
            print(f"{j+1:4}: {lines[j]}")
            shown.add(j)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/"
    "bearing-uav-90k-city-archives"
)

meta = root / (
    "Bearing_UAV_90K/"
    "c4m_254k_96bc_b15_s100_v3d/"
    "metadata/metadata.csv"
)

df = pd.read_csv(meta)
sg = df[df["target_path"].str.contains("_37bc_")].copy()

for _, r in sg.head(5).iterrows():
    bx, by = int(r.block_x), int(r.block_y)

    # 메타데이터의 정규화 좌표를 지도 픽셀 좌표로 환산
    real_x = bx * 256 + 256 + r.x_norm * 128
    real_y = by * 256 + 256 + r.y_norm * 128

    # nav_run이 사용하는 블록 중심
    center_x = bx * 256 + 256
    center_y = by * 256 + 256

    print(
        f"{r.target_path.split('/')[-1]} | "
        f"block=({bx},{by}) | "
        f"norm=({r.x_norm:.4f},{r.y_norm:.4f}) | "
        f"real=({real_x:.2f},{real_y:.2f}) | "
        f"center=({center_x},{center_y}) | "
        f"offset=({real_x-center_x:.2f},{real_y-center_y:.2f})"
    )
```

```python
from pathlib import Path

p = Path(
    "/kaggle/input/datasets/jimin562/c5-trone-imatch-evaluation/imatch1/nav.py"
)
lines = p.read_text(encoding="utf-8").splitlines()

def show_section(start_text, end_text=None):
    start = next(
        (i for i, line in enumerate(lines) if start_text in line),
        None
    )
    if start is None:
        print(f"\n[찾지 못함] {start_text}")
        return

    end = len(lines)
    if end_text:
        found = next(
            (i for i in range(start + 1, len(lines))
             if end_text in lines[i]),
            None
        )
        if found is not None:
            end = found

    print(f"\n===== {start_text} =====")
    for i in range(start, end):
        print(f"{i+1:4}: {lines[i]}")

show_section("class Nearest3D", "class NoMatch")
show_section("def nav_run", None)
```

```python
import sys
import tarfile
import json
import torch
from pathlib import Path

root = Path(
    "/kaggle/input/datasets/jimin562/capstone1/"
    "capstone1-main/capstone1-main"
)

sys.path.insert(0, str(root))

from cvphr.models.posaglreg.models import load_config_and_model

archive = "/kaggle/input/datasets/jimin562/capstone1/b1_retrained.tgz"

base = (
    "c4ma/phr5_d96100_3d_ho37bc_b16_l0.5_e100_gNone_"
    "20261005_152953"
)

# 임시 디렉터리에 설정 파일만 추출
temp = Path("/kaggle/working/b1_load_test")
temp.mkdir(exist_ok=True, parents=True)

with tarfile.open(archive, "r:gz") as tar:
    for name in ["training_configure.json"]:
        member = tar.getmember(f"{base}/{name}")
        member.name = name
        tar.extract(member, path=temp)

# 모델 생성
cls, kwargs = load_config_and_model(str(temp))
model = cls(**kwargs)

# 가중치 로드
with tarfile.open(archive, "r:gz") as tar:
    ckpt = torch.load(
        tar.extractfile(f"{base}/best_model.pth"),
        map_location="cpu",
        weights_only=False
    )

result = model.load_state_dict(
    ckpt["model_state_dict"],
    strict=True
)

print("모델 클래스:", cls.__name__)
print("모델 파라미터 수:", sum(p.numel() for p in model.parameters()))
print("가중치 로딩:", result)
print("B1 모델 로딩 테스트 성공")
```

```python
import torch
from pathlib import Path
from cvphr.models.posaglreg.models import load_config_and_model

def load_b1(model_dir):
    cls, kwargs = load_config_and_model(str(model_dir))
    model = cls(**kwargs)

    checkpoint = torch.load(
        model_dir / "best_model.pth",
        map_location="cpu",
        weights_only=False
    )

    result = model.load_state_dict(
        checkpoint["model_state_dict"],
        strict=True
    )

    model = model.to("cuda").eval()

    print(f"\n[성공] {model_dir.name}")
    print("파라미터:", sum(p.numel() for p in model.parameters()))
    print("Missing:", len(result.missing_keys))
    print("Unexpected:", len(result.unexpected_keys))

    return model

b1_singapore = load_b1(Path("/kaggle/working/b1_singapore"))
b1_sf = load_b1(Path("/kaggle/working/b1_sf"))

print("\n두 도시 B1 모델 로딩 완료")
```

```python
import torch
import cv2
import numpy as np

# 기존에 확인한 Singapore 샘플 사용
model = b1_singapore
model.eval()

# 타일 4장 + UAV 영상
tile_tensors = []
for col in ["p1_path", "p2_path", "p3_path", "p4_path"]:
    img = cv2.imread(str(BASE / row[col]))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    tile_tensors.append(to_tensor(img))

uav_img = cv2.imread(str(BASE / row["target_path"]))
uav_img = cv2.cvtColor(uav_img, cv2.COLOR_BGR2RGB)
uav_tensor = to_tensor(uav_img)

tiles = torch.stack(tile_tensors).to(DEV)
uav = uav_tensor.unsqueeze(0).to(DEV)

with torch.no_grad():
    # 방식 A: 원본 이미지 5장을 직접 입력
    patches = torch.cat([tiles, uav], dim=0).unsqueeze(0)
    pos_a, dir_a = model(patches)

    # 방식 B: 위성 타일은 backbone 특징맵으로 미리 변환
    tile_feats = model.backbone(tiles)
    tile_feats = tile_feats.unsqueeze(0)

    pos_b, dir_b = model.forward_tile_feats(uav, tile_feats)

print("Backbone tile features:", tuple(tile_feats.shape))
print("일반 forward 위치:", pos_a.cpu().numpy())
print("캐시 방식 위치:", pos_b.cpu().numpy())
print("위치 최대 차이:", (pos_a - pos_b).abs().max().item())
print()
print("일반 forward 방향:", dir_a.cpu().numpy())
print("캐시 방식 방향:", dir_b.cpu().numpy())
print("방향 최대 차이:", (dir_a - dir_b).abs().max().item())
```

```python
from scipy.spatial import cKDTree
import numpy as np

for city_code in ["37bc", "38bc"]:
    sub = md[
        md["p1_path"].astype(str).str.contains(
            city_code, regex=False, na=False
        )
    ].copy()

    xy = np.column_stack([
        sub["block_x"].to_numpy() * 256 + 256
        + sub["x_norm"].to_numpy() * 128,

        sub["block_y"].to_numpy() * 256 + 256
        + sub["y_norm"].to_numpy() * 128
    ])

    tree = cKDTree(xy)
    distances, indices = tree.query(xy, k=2)

    # 자기 자신을 제외한 최근접 샘플 거리
    nn_px = distances[:, 1]
    nn_m = nn_px * 0.25

    print(f"\n===== {city_code} =====")
    print(f"중앙값: {np.median(nn_m):.2f} m")
    print(f"평균: {np.mean(nn_m):.2f} m")
    print(f"95 백분위: {np.percentile(nn_m, 95):.2f} m")
    print(f"최대: {np.max(nn_m):.2f} m")
    print(f"15m 이내: {(nn_m <= 15).mean()*100:.1f}%")
    print(f"30m 이내: {(nn_m <= 30).mean()*100:.1f}%")
```

```python
import numpy as np
import cv2
from pathlib import Path
from scipy.spatial import cKDTree

class B1Data3D:
    def __init__(self, metadata, city_code, base_dir):
        self.base = Path(base_dir)

        self.df = metadata[
            metadata["p1_path"].astype(str).str.contains(
                city_code, regex=False, na=False
            )
        ].copy().reset_index(drop=True)

        # 시뮬레이션 지도 좌표로 변환
        self.df["px"] = (
            self.df["block_x"] * 256
            + 256 + self.df["x_norm"] * 128
        )
        self.df["py"] = (
            self.df["block_y"] * 256
            + 256 + self.df["y_norm"] * 128
        )

        self.xy = self.df[["px", "py"]].to_numpy(
            dtype=np.float32
        )
        self.tree = cKDTree(self.xy)

        self.tile_cache = {}
        self.target_cache = {}

        print(
            f"{city_code}: {len(self.df)} samples, "
            f"{self.df[['block_x','block_y']].drop_duplicates().shape[0]} blocks"
        )

    def nearest_row(self, x, y):
        distance, idx = self.tree.query([x, y], k=1)
        return self.df.iloc[int(idx)], float(distance)

    def target(self, row):
        path = str(row["target_path"])

        if path not in self.target_cache:
            img = cv2.imread(str(self.base / path))
            if img is None:
                raise FileNotFoundError(str(self.base / path))

            self.target_cache[path] = cv2.cvtColor(
                img, cv2.COLOR_BGR2RGB
            )

        return self.target_cache[path]

    def tiles(self, row):
        key = (int(row["block_x"]), int(row["block_y"]))

        if key not in self.tile_cache:
            paths = [
                row["p1_path"], row["p2_path"],
                row["p3_path"], row["p4_path"]
            ]

            imgs = []
            for path in paths:
                img = cv2.imread(str(self.base / path))
                if img is None:
                    raise FileNotFoundError(str(self.base / path))

                imgs.append(
                    cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                )

            self.tile_cache[key] = imgs

        return self.tile_cache[key]


BASE = (
    "/kaggle/input/datasets/jimin562/"
    "bearing-uav-90k-city-archives"
)

data_sg = B1Data3D(md, "37bc", BASE)
data_sf = B1Data3D(md, "38bc", BASE)

# 실제 데이터 공급 테스트
for name, data in [("Singapore", data_sg), ("SF", data_sf)]:
    row, distance = data.nearest_row(
        data.xy[0, 0], data.xy[0, 1]
    )

    print(f"\n{name}")
    print("최근접 거리:", distance, "px")
    print("UAV 영상:", data.target(row).shape)
    print("위성 타일:", [im.shape for im in data.tiles(row)])
    print("theta:", float(row["theta"]))
    print("선택 좌표:", row["px"], row["py"])
```

```python
import torch
import numpy as np

class B1Loc:
    def __init__(self, model, data):
        self.m = model
        self.data = data
        self.tile_feature_cache = {}

        # 블록별 대표 메타데이터 행
        self.block_rows = (
            data.df.drop_duplicates(["block_x", "block_y"])
            .set_index(["block_x", "block_y"])
        )

    def block_features(self, bx, by):
        key = (int(bx), int(by))

        if key not in self.tile_feature_cache:
            row = self.block_rows.loc[key]

            tiles = self.data.tiles(row)
            x = torch.stack([
                to_tensor(im) for im in tiles
            ]).to(DEV)

            with torch.no_grad():
                self.tile_feature_cache[key] = enc(
                    self.m, x
                ).detach()

        return self.tile_feature_cache[key]

    def __call__(self, crops, headings, blocks):
        # UAV 영상 특징
        uav = torch.stack([
            to_tensor(c) for c in crops
        ]).to(DEV)

        # 블록별 위성 타일 특징
        nb = torch.stack([
            self.block_features(bx, by)
            for bx, by in blocks
        ])

        with torch.no_grad():
            fi = enc(self.m, uav)
            outputs = head(self.m, fi, nb)

        self.last_direction = (
            outputs[1].detach().cpu().numpy()
        )

        return outputs[0].detach().cpu().numpy()


# Singapore B1 어댑터
b1_loc_sg = B1Loc(b1_singapore, data_sg)

# 첫 번째 샘플로 단일 추론 테스트
sample = data_sg.df.iloc[0]
bx = int(sample["block_x"])
by = int(sample["block_y"])

uav_img = data_sg.target(sample)

pred = b1_loc_sg(
    [uav_img],
    [float(sample["theta"])],
    [(bx, by)]
)

print("위치 예측:", pred)
print("방향 예측:", b1_loc_sg.last_direction)
print("위치 출력 shape:", pred.shape)
print("타일 특징 캐시:", len(b1_loc_sg.tile_feature_cache))
```

```python
# B1Data3D.tiles 메서드 수정
def _tiles_fixed(self, row):
    # block_x, block_y가 일반 컬럼 또는 인덱스인 경우 모두 대응
    if "block_x" in row.index:
        bx = int(row["block_x"])
        by = int(row["block_y"])
    else:
        bx = int(row.name[0])
        by = int(row.name[1])

    key = (bx, by)

    if key not in self.tile_cache:
        paths = [
            row["p1_path"], row["p2_path"],
            row["p3_path"], row["p4_path"]
        ]

        imgs = []
        for path in paths:
            img = cv2.imread(str(self.base / path))
            if img is None:
                raise FileNotFoundError(str(self.base / path))

            imgs.append(
                cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            )

        self.tile_cache[key] = imgs

    return self.tile_cache[key]


B1Data3D.tiles = _tiles_fixed

print("tiles 메서드 수정 완료")
```

```python
import numpy as np

def inspect_direction(name, loc, sample):
    direction = loc.last_direction[0]

    pred_angle = np.degrees(
        np.arctan2(direction[1], direction[0])
    )
    pred_norm = np.linalg.norm(direction)
    gt_theta = float(sample["theta"])

    print(f"===== {name} =====")
    print(f"정답 theta: {gt_theta:.2f}°")
    print(f"예측 방향 벡터: {direction}")
    print(f"예측 벡터 크기: {pred_norm:.4f}")
    print(f"예측 벡터 각도(atan2): {pred_angle:.2f}°")

inspect_direction("Singapore", b1_loc_sg, data_sg.df.iloc[0])
inspect_direction("San Francisco", b1_loc_sf, data_sf.df.iloc[0])
```

```python
import numpy as np

def check_b1_direction(name, loc, sample):
    pred = loc.last_direction[0]
    gt = np.array([
        sample["x_cosa"],
        sample["y_sina"]
    ], dtype=np.float32)

    pred_angle = np.degrees(np.arctan2(pred[1], pred[0]))
    gt_angle = np.degrees(np.arctan2(gt[1], gt[0]))

    cosine = np.dot(pred, gt) / (
        np.linalg.norm(pred) * np.linalg.norm(gt) + 1e-8
    )

    error = abs((pred_angle - gt_angle + 180) % 360 - 180)

    print(f"===== {name} =====")
    print("정답 벡터:", gt)
    print("예측 벡터:", pred)
    print(f"정답 각도: {gt_angle:.2f}°")
    print(f"예측 각도: {pred_angle:.2f}°")
    print(f"코사인 유사도: {cosine:.4f}")
    print(f"각도 오차: {error:.2f}°")

check_b1_direction("Singapore", b1_loc_sg, data_sg.df.iloc[0])
check_b1_direction("San Francisco", b1_loc_sf, data_sf.df.iloc[0])
```

```python
import numpy as np
from tqdm.auto import tqdm

def evaluate_b1_directions(name, loc, data, n=100):
    errors = []
    cosines = []

    count = min(n, len(data.df))

    for i in tqdm(range(count), desc=name):
        row = data.df.iloc[i]

        uav_img = data.target(row)
        block = (
            int(row["block_x"]),
            int(row["block_y"])
        )

        loc(
            [uav_img],
            [float(row["theta"])],
            [block]
        )

        pred = np.asarray(
            loc.last_direction[0],
            dtype=np.float64
        )

        gt = np.array(
            [row["x_cosa"], row["y_sina"]],
            dtype=np.float64
        )

        pred_norm = np.linalg.norm(pred)
        gt_norm = np.linalg.norm(gt)

        if pred_norm < 1e-8 or gt_norm < 1e-8:
            continue

        cosine = np.dot(pred, gt) / (pred_norm * gt_norm)
        cosine = np.clip(cosine, -1.0, 1.0)

        error = np.degrees(np.arccos(cosine))

        errors.append(error)
        cosines.append(cosine)

    errors = np.asarray(errors)
    cosines = np.asarray(cosines)

    print(f"\n===== {name}: 방향 평가 =====")
    print(f"평가 표본: {len(errors)}")
    print(f"평균 각도 오차: {errors.mean():.2f}°")
    print(f"중앙값 각도 오차: {np.median(errors):.2f}°")
    print(f"90% 분위 각도 오차: {np.percentile(errors, 90):.2f}°")
    print(f"평균 코사인 유사도: {cosines.mean():.4f}")
    print(f"30° 이내 비율: {(errors <= 30).mean() * 100:.1f}%")
    print(f"45° 이내 비율: {(errors <= 45).mean() * 100:.1f}%")

    return errors, cosines


b1_sg_errors, b1_sg_cosines = evaluate_b1_directions(
    "Singapore", b1_loc_sg, data_sg, n=100
)

b1_sf_errors, b1_sf_cosines = evaluate_b1_directions(
    "San Francisco", b1_loc_sf, data_sf, n=100
)
```

```python
import numpy as np
from tqdm.auto import tqdm

def evaluate_b1_positions(name, loc, data, n=100):
    errors = []
    count = min(n, len(data.df))

    for i in tqdm(range(count), desc=name):
        row = data.df.iloc[i]

        uav_img = data.target(row)
        block = (
            int(row["block_x"]),
            int(row["block_y"])
        )

        pred = loc(
            [uav_img],
            [float(row["theta"])],
            [block]
        )[0]

        gt = np.array(
            [row["x_norm"], row["y_norm"]],
            dtype=np.float64
        )

        # B1 상대 위치 단위: 1 unit = 32m
        error_m = np.linalg.norm(
            (np.asarray(pred, dtype=np.float64) - gt) * 32.0
        )

        errors.append(error_m)

    errors = np.asarray(errors)

    print(f"\n===== {name}: 위치 평가 =====")
    print(f"평가 표본: {len(errors)}")
    print(f"평균 위치 오차: {errors.mean():.2f} m")
    print(f"중앙 위치 오차: {np.median(errors):.2f} m")
    print(f"90% 분위 위치 오차: {np.percentile(errors, 90):.2f} m")
    print(f"10m 이내 비율: {(errors <= 10).mean() * 100:.1f}%")
    print(f"20m 이내 비율: {(errors <= 20).mean() * 100:.1f}%")

    return errors


b1_sg_pos_errors = evaluate_b1_positions(
    "Singapore", b1_loc_sg, data_sg, n=100
)

b1_sf_pos_errors = evaluate_b1_positions(
    "San Francisco", b1_loc_sf, data_sf, n=100
)
```

```python
import cv2
import numpy as np

keys = list(b1_loc_sg.block_rows.index[:50])
errs, preds, gts = [], [], []

for key in keys:
    row = b1_loc_sg.block_rows.loc[key]
    path = b1_loc_sg.data.base / row["target_path"]
    img = cv2.cvtColor(cv2.imread(str(path)), cv2.COLOR_BGR2RGB)
    p = b1_loc_sg([img], [90.0], [key])[0]
    g = np.array([row["x_norm"], row["y_norm"]], dtype=float)
    preds.append(p); gts.append(g)
    errs.append(np.linalg.norm(p - g))

preds, gts, errs = np.array(preds), np.array(gts), np.array(errs)

print("n =", len(errs))
print("평균 오차(norm):", errs.mean(), " 중앙값:", np.median(errs))
print("평균 오차(px@128):", errs.mean() * 128)
print("예측 mean/std:", preds.mean(0), preds.std(0))
print("정답 mean/std:", gts.mean(0), gts.std(0))
print("상관계수 x:", np.corrcoef(preds[:, 0], gts[:, 0])[0, 1])
print("상관계수 y:", np.corrcoef(preds[:, 1], gts[:, 1])[0, 1])
```

```python
import numpy as np, cv2

df = b1_loc_sg.block_rows
print("x_norm min/max:", df["x_norm"].min(), df["x_norm"].max())
print("y_norm min/max:", df["y_norm"].min(), df["y_norm"].max())
print("컬럼:", list(df.columns))   # split/train/val 같은 컬럼 확인

# 무작위 100개 (앞쪽 편향 제거)
rng = np.random.default_rng(0)
keys = [df.index[i] for i in rng.choice(len(df), size=min(100, len(df)), replace=False)]

P, G = [], []
for k in keys:
    r = df.loc[k]
    img = cv2.cvtColor(cv2.imread(str(b1_loc_sg.data.base / r["target_path"])), cv2.COLOR_BGR2RGB)
    P.append(b1_loc_sg([img], [90.0], [k])[0])
    G.append([r["x_norm"], r["y_norm"]])
P, G = np.array(P), np.array(G, dtype=float)

err = np.linalg.norm(P - G, axis=1)
base = np.linalg.norm(G - G.mean(0), axis=1)   # '평균만 찍는' 기준선
print("모델 평균 오차:", err.mean(), "| 평균 예측 기준선:", base.mean())

# 선형 보정 후 오차 (스케일/오프셋 문제인지)
A = np.c_[P, np.ones(len(P))]
W, *_ = np.linalg.lstsq(A, G, rcond=None)
print("선형 보정 후 오차:", np.linalg.norm(A @ W - G, axis=1).mean())

# x/y 뒤바뀜 가설
print("swap 시 오차:", np.linalg.norm(P[:, ::-1] - G, axis=1).mean())
```

```python
from imatch1.nav import make_traj, nav_run, ODO, Nearest3D

CAPSTONE_ROOT = str(DATA_ROOT.parent)

# 짧은 시험 경로: 1개, 30m, seed 7
starts, goals = make_traj(
    n_traj=1,
    dist_m=30,
    seed=7
)

print("시작점:", starts[0])
print("목표점:", goals[0])

# 도시별 테스트
for code, name, loc in [
    ("37bc", "Singapore", b1_loc_sg),
    ("38bc", "San Francisco", b1_loc_sf),
]:
    frames = Nearest3D(CAPSTONE_ROOT, f"3d_ho{code}")
    log = []

    success, reason, errors = nav_run(
        loc=loc,
        frames=frames,
        starts=starts,
        goals=goals,
        decode_px=128,
        step_m=15,
        arrive_m=20,
        n_block=15,
        odo=ODO["vio"],
        seed=7,
        log=log,
    )

    print(f"\n[{name} / {code}]")
    print("성공:", success)
    print("종료 이유:", reason)
    print("궤적별 오차 중앙값:", errors)
    print("기록된 걸음 수:", len(log))
    print("첫 걸음 기록:", log[0] if log else None)
```

**싱가포르 B1 본 실험**

```python
import pandas as pd
import numpy as np
from imatch1.nav import make_traj, nav_run, ODO, Nearest3D

# 두 도시가 공유할 동일한 경로 생성
nav_starts, nav_goals = make_traj(
    n_traj=32,
    dist_m=1000,
    seed=7
)

print("경로 수:", len(nav_starts))
print("경로 길이 평균(m):",
      np.mean(np.linalg.norm(nav_goals - nav_starts, axis=1)) / 4)

# 싱가포르 B1 본 실험
sg_frames = Nearest3D(CAPSTONE_ROOT, "3d_ho37bc")
sg_log = []

sg_success, sg_reason, sg_errors = nav_run(
    loc=b1_loc_sg,
    frames=sg_frames,
    starts=nav_starts,
    goals=nav_goals,
    decode_px=128,
    step_m=15,
    arrive_m=20,
    n_block=15,
    odo=ODO["vio"],
    seed=7,
    log=sg_log,
)

sg_results = pd.DataFrame(sg_log)
sg_results.to_csv(
    "/kaggle/working/B1_3d_37bc_nav_log.csv",
    index=False
)

print("\n===== B1 Singapore 3D Navigation =====")
print("성공:", int(sg_success.sum()), "/", len(sg_success))
print("성공률:", round(sg_success.mean() * 100, 2), "%")
print("종료 이유:")
print(pd.Series(sg_reason).value_counts())
print("궤적별 오차 중앙값 평균:",
      round(float(np.mean(sg_errors)), 2), "m")
print("총 기록 걸음:", len(sg_log))
print("로그 저장 완료")
```

```python

import pandas as pd
import numpy as np

# Singapore B1 궤적별 성공·실패 결과
sg_b1_traj = pd.DataFrame({
    "traj": np.arange(len(sg_success)),
    "success": np.asarray(sg_success, dtype=bool),
    "reason": np.asarray(sg_reason, dtype=str),
    "median_error_m": np.asarray(sg_errors, dtype=float),
})

# 파일 저장
traj_path = "/kaggle/working/SG_B1_3D_VIO_seed7_traj.csv"
sg_b1_traj.to_csv(traj_path, index=False)

# 결과 확인
print("===== Singapore B1 궤적별 결과 =====")
print(sg_b1_traj.to_string(index=False))

print("\n===== 성공·실패 집계 =====")
print(sg_b1_traj["reason"].value_counts().to_string())

print("\n성공률:",
      round(sg_b1_traj["success"].mean() * 100, 2), "%")

print("\n저장 완료:", traj_path)
```

```python
import pandas as pd
import numpy as np

# 궤적별 오차 통계
summary = sg_results.groupby("traj").agg(
    steps=("step", "count"),
    median_err_m=("err_m", "median"),
    mean_err_m=("err_m", "mean"),
    p90_err_m=("err_m", lambda x: x.quantile(0.9)),
    max_err_m=("err_m", "max"),
    last_err_m=("err_m", "last"),
    outside_steps=("outside_block", "sum"),
).reset_index()

summary["success"] = sg_success[summary["traj"].to_numpy()]
summary["reason"] = sg_reason[summary["traj"].to_numpy()]

print("===== 궤적별 오차가 큰 순서 =====")
display(summary.sort_values("median_err_m", ascending=False).head(10))

print("\n===== 종료 이유별 오차 =====")
display(
    summary.groupby("reason").agg(
        trajectories=("traj", "count"),
        median_error=("median_err_m", "median"),
        mean_error=("median_err_m", "mean"),
        median_steps=("steps", "median"),
    )
)

# 오차가 가장 큰 궤적의 걸음별 기록
worst_id = int(summary.loc[summary["median_err_m"].idxmax(), "traj"])
worst = sg_results[sg_results["traj"] == worst_id].sort_values("step")

print(f"\n===== 오차가 가장 큰 궤적: {worst_id} =====")
display(
    worst[
        ["step", "real_x", "real_y", "est_x", "est_y",
         "err_m", "raw_err_m", "block_x", "block_y",
         "outside_block", "unknown", "used"]
    ].head(15)
)

print("\n===== 마지막 5걸음 =====")
display(worst.tail(5))
```

```python
# 기존 싱가포르 경로와 동일한 조건
sg_oracle_log = []

sg_oracle_success, sg_oracle_reason, sg_oracle_errors = nav_run(
    loc=b1_loc_sg,
    frames=sg_frames,
    starts=nav_starts,
    goals=nav_goals,
    decode_px=128,
    oracle_block=True,  # 실제 위치로 지도 블록 선택
    step_m=15,
    arrive_m=20,
    n_block=15,
    odo=ODO["vio"],
    seed=7,
    log=sg_oracle_log,
)

sg_oracle_results = pd.DataFrame(sg_oracle_log)

print("===== B1 Singapore: Oracle Block 진단 =====")
print(
    "성공률:",
    round(sg_oracle_success.mean() * 100, 2),
    "%"
)
print(
    "성공:",
    int(sg_oracle_success.sum()),
    "/",
    len(sg_oracle_success)
)
print("종료 이유:")
print(pd.Series(sg_oracle_reason).value_counts())
print(
    "궤적별 오차 중앙값 평균:",
    round(float(np.mean(sg_oracle_errors)), 2),
    "m"
)
print(
    "outside_block 비율:",
    round(sg_oracle_results["outside_block"].mean() * 100, 2),
    "%"
)
```

```python
# 실제 위치로 계산한 정답 블록
correct_bx = np.clip(
    np.round((sg_results["real_x"] - 256) / 256), 0, 14
).astype(int)

correct_by = np.clip(
    np.round((sg_results["real_y"] - 256) / 256), 0, 14
).astype(int)

# 일반 B1이 선택한 블록과 비교
wrong_block = (
    (sg_results["block_x"].to_numpy() != correct_bx) |
    (sg_results["block_y"].to_numpy() != correct_by)
)

print("===== 일반 B1 블록 선택 진단 =====")
print("전체 걸음:", len(sg_results))
print("잘못 선택한 블록:", int(wrong_block.sum()))
print("블록 선택 오류율:", round(wrong_block.mean() * 100, 2), "%")

# 궤적별 최초 블록 선택 오류
tmp = sg_results.copy()
tmp["wrong_block"] = wrong_block

first_wrong = (
    tmp[tmp["wrong_block"]]
    .groupby("traj")["step"]
    .min()
)

print("\n블록 선택 오류가 발생한 궤적:", len(first_wrong), "/ 32")
print("최초 오류 걸음 중앙값:", first_wrong.median())
print("\n최초 오류 걸음 예시:")
print(first_wrong.head(10))
```

```python
# 32개 시작 좌표와 Nearest3D 반환 좌표의 블록 비교

def get_block(x, y):
    bx = int(np.clip(np.round((x - 256) / 256), 0, 14))
    by = int(np.clip(np.round((y - 256) / 256), 0, 14))
    return bx, by

start_check = []

for i, (sx, sy) in enumerate(nav_starts):
    _, fx, fy = sg_frames(sx, sy, 90.0)

    input_block = get_block(sx, sy)
    frame_block = get_block(fx, fy)

    start_check.append({
        "traj": i,
        "input_bx": input_block[0],
        "input_by": input_block[1],
        "frame_bx": frame_block[0],
        "frame_by": frame_block[1],
        "same_block": input_block == frame_block,
        "frame_dx": fx - sx,
        "frame_dy": fy - sy,
    })

start_check = pd.DataFrame(start_check)

print("===== 시작점 블록 일치 검사 =====")
print("전체:", len(start_check))
print("일치:", int(start_check["same_block"].sum()))
print("불일치:", int((~start_check["same_block"]).sum()))

print("\n===== 불일치 사례 =====")
print(
    start_check[~start_check["same_block"]]
    .round(2)
    .to_string(index=False)
)
```

```python
import numpy as np

# 기존 항법 시작점 32개에서 첫 번째 프레임 생성
test_crops = []
test_blocks = []

for i in range(len(nav_starts)):
    x, y = nav_starts[i]

    bx = int(np.clip(np.round((x - 256) / 256), 0, 14))
    by = int(np.clip(np.round((y - 256) / 256), 0, 14))

    crop, fx, fy = sg_frames(x, y, 90.0)

    test_crops.append(crop)
    test_blocks.append((bx, by))

# B1 위치 출력
test_pos = b1_loc_sg(test_crops, np.full(len(test_crops), 90.0), test_blocks)

test_pos = np.asarray(test_pos)

print("===== B1 위치 출력 통계 =====")
print("shape:", test_pos.shape)

for axis, values in zip(["x_norm", "y_norm"], test_pos.T):
    print(
        f"{axis}: "
        f"mean={values.mean():.4f}, "
        f"std={values.std():.4f}, "
        f"min={values.min():.4f}, "
        f"max={values.max():.4f}"
    )

print("\n===== 첫 10개 출력 =====")
print(np.round(test_pos[:10], 4))

print("\n===== 정규화 범위 초과 =====")
print("x:", np.sum(np.abs(test_pos[:, 0]) > 1))
print("y:", np.sum(np.abs(test_pos[:, 1]) > 1))
```

```python
import numpy as np
import pandas as pd

rows = []

for i in range(len(nav_starts)):
    x, y = nav_starts[i]

    crop, fx, fy = sg_frames(x, y, 90.0)

    bx, by = test_blocks[i]
    center = np.array([
        bx * 256 + 256,
        by * 256 + 256
    ])

    pred = center + test_pos[i] * 128
    real = np.array([fx, fy])

    rows.append({
        "traj": i,
        "block_x": bx,
        "block_y": by,
        "real_x": fx,
        "real_y": fy,
        "pred_x": pred[0],
        "pred_y": pred[1],
        "err_m": np.linalg.norm(pred - real) / PX,
        "outside_block": int(
            np.abs(real - center).max() > 128
        )
    })

first_step = pd.DataFrame(rows)

print("===== 첫 프레임 위치 추정 결과 =====")
print("표본 수:", len(first_step))
print("평균 오차:", round(first_step.err_m.mean(), 2), "m")
print("중앙 오차:", round(first_step.err_m.median(), 2), "m")
print("90% 오차:", round(first_step.err_m.quantile(.9), 2), "m")
print("10m 이내:", (first_step.err_m <= 10).sum(), "/ 32")
print("20m 이내:", (first_step.err_m <= 20).sum(), "/ 32")
print("블록 중심 ±128px 밖:", first_step.outside_block.sum(), "/ 32")

display(
    first_step.sort_values("err_m", ascending=False).head(10)
)
```

```python
import numpy as np
import pandas as pd

PX = 0.25  # 미터/픽셀
rows = []

for i in range(len(nav_starts)):
    x, y = nav_starts[i]

    crop, fx, fy = sg_frames(x, y, 90.0)

    bx, by = test_blocks[i]
    center = np.array([
        bx * 256 + 256,
        by * 256 + 256
    ])

    pred = center + test_pos[i] * 128
    real = np.array([fx, fy])

    err_m = np.linalg.norm(pred - real) / PX

    rows.append({
        "traj": i,
        "block_x": bx,
        "block_y": by,
        "real_x": fx,
        "real_y": fy,
        "pred_x": pred[0],
        "pred_y": pred[1],
        "err_m": err_m,
        "outside_block": int(
            np.abs(real - center).max() > 128
        )
    })

first_step = pd.DataFrame(rows)

print("===== 첫 프레임 위치 추정 결과 =====")
print("표본 수:", len(first_step))
print("평균 오차:", round(first_step.err_m.mean(), 2), "m")
print("중앙 오차:", round(first_step.err_m.median(), 2), "m")
print("90% 오차:", round(first_step.err_m.quantile(.9), 2), "m")
print("10m 이내:", (first_step.err_m <= 10).sum(), "/ 32")
print("20m 이내:", (first_step.err_m <= 20).sum(), "/ 32")
print("블록 중심 ±128px 밖:", first_step.outside_block.sum(), "/ 32")

display(first_step.sort_values("err_m", ascending=False).head(10))
```

```python
PX = 0.25  # 1픽셀 = 0.25m

first_step["err_px"] = np.sqrt(
    (first_step["pred_x"] - first_step["real_x"])**2
    + (first_step["pred_y"] - first_step["real_y"])**2
)

first_step["err_m"] = first_step["err_px"] * PX

print("===== 수정된 첫 프레임 위치 추정 결과 =====")
print("표본 수:", len(first_step))
print("평균 오차:", round(first_step.err_m.mean(), 2), "m")
print("중앙 오차:", round(first_step.err_m.median(), 2), "m")
print("90% 오차:", round(first_step.err_m.quantile(.9), 2), "m")
print("10m 이내:", (first_step.err_m <= 10).sum(), "/ 32")
print("20m 이내:", (first_step.err_m <= 20).sum(), "/ 32")

display(first_step.sort_values("err_m", ascending=False).head(10))
```

```python
# 항법 시작 좌표 추가
first_step["start_x"] = [p[0] for p in nav_starts]
first_step["start_y"] = [p[1] for p in nav_starts]

# 세 종류의 오차 계산 (미터)
first_step["frame_offset_m"] = np.sqrt(
    (first_step["real_x"] - first_step["start_x"])**2
    + (first_step["real_y"] - first_step["start_y"])**2
) * PX

first_step["pred_to_frame_m"] = np.sqrt(
    (first_step["pred_x"] - first_step["real_x"])**2
    + (first_step["pred_y"] - first_step["real_y"])**2
) * PX

first_step["pred_to_start_m"] = np.sqrt(
    (first_step["pred_x"] - first_step["start_x"])**2
    + (first_step["pred_y"] - first_step["start_y"])**2
) * PX

print("===== 첫 프레임 오차 원인 분리 =====")

for col, name in [
    ("frame_offset_m", "시작 좌표 ↔ 선택 프레임"),
    ("pred_to_frame_m", "선택 프레임 ↔ B1 예측"),
    ("pred_to_start_m", "시작 좌표 ↔ B1 예측"),
]:
    print(
        f"{name}: "
        f"평균 {first_step[col].mean():.2f}m / "
        f"중앙값 {first_step[col].median():.2f}m"
    )

print("\n===== 오차가 큰 표본 10개 =====")
display(
    first_step.sort_values(
        "pred_to_frame_m", ascending=False
    )[[
        "traj",
        "frame_offset_m",
        "pred_to_frame_m",
        "pred_to_start_m",
        "outside_block"
    ]].head(10)
)
```

```python
# 시작 좌표 기준 X/Y 오차 (미터)
first_step["dx_m"] = (
    first_step["pred_x"] - first_step["start_x"]
) * PX

first_step["dy_m"] = (
    first_step["pred_y"] - first_step["start_y"]
) * PX

print("===== B1 첫 위치 추정 방향별 오차 =====")

for axis in ["dx_m", "dy_m"]:
    v = first_step[axis]

    print(f"\n{axis}")
    print("평균 편향:", round(v.mean(), 2), "m")
    print("표준편차:", round(v.std(), 2), "m")
    print("최솟값:", round(v.min(), 2), "m")
    print("최댓값:", round(v.max(), 2), "m")
    print("절대 오차 중앙값:", round(v.abs().median(), 2), "m")

print("\n===== 궤적별 방향 오차 =====")
display(
    first_step[["traj", "dx_m", "dy_m", "err_m"]]
    .sort_values("err_m", ascending=False)
    .head(10)
)
```

```python
print("===== 블록별 첫 프레임 오차 =====")

block_stats = (
    first_step
    .groupby(["block_x", "block_y"])
    .agg(
        n=("traj", "count"),
        mean_error=("err_m", "mean"),
        mean_dx=("dx_m", "mean"),
        mean_dy=("dy_m", "mean")
    )
    .reset_index()
    .sort_values("mean_error", ascending=False)
)

display(block_stats)

print("\n===== 블록별 표본 수 =====")
print(block_stats["n"].value_counts().sort_index())
```

```python
print("===== 지도 좌표와 B1 오차의 상관관계 =====")

for coord in ["start_x", "start_y"]:
    for err in ["err_m", "dx_m", "dy_m"]:
        corr = np.corrcoef(
            first_step[coord],
            first_step[err]
        )[0, 1]

        print(f"{coord} vs {err}: r={corr:.3f}")

print("\n===== 지도 위치 구간별 오차 =====")

first_step["x_region"] = pd.cut(
    first_step["start_x"],
    bins=4,
    labels=["서쪽", "중서쪽", "중동쪽", "동쪽"]
)

first_step["y_region"] = pd.cut(
    first_step["start_y"],
    bins=4,
    labels=["북쪽", "중북쪽", "중남쪽", "남쪽"]
)

print("\n[X 방향 구간별]")
display(
    first_step.groupby("x_region", observed=True)
    .agg(
        n=("traj", "count"),
        mean_error=("err_m", "mean"),
        mean_dx=("dx_m", "mean"),
        mean_dy=("dy_m", "mean")
    )
)

print("\n[Y 방향 구간별]")
display(
    first_step.groupby("y_region", observed=True)
    .agg(
        n=("traj", "count"),
        mean_error=("err_m", "mean"),
        mean_dx=("dx_m", "mean"),
        mean_dy=("dy_m", "mean")
    )
)
```

```python
import torch
import numpy as np

from cvphr.models.corrmatch import CorrMatchNet, north_up

# 1. CorrMatchNet 로드 (학습된 보정 설정 유지)
corr_model = CorrMatchNet.load(
    corr_model_dir,
    device=DEV
)

print("모델 로드 완료:", type(corr_model).__name__)

# 2. B1 데이터 타일을 사용하는 어댑터
class CorrNavLoc:
    def __init__(self, model, data, block_rows):
        self.m = model
        self.data = data
        self.block_rows = block_rows
        self.cache = {}

    @torch.no_grad()
    def __call__(self, crops, headings, blocks):
        tf_batch = []

        for bx, by in blocks:
            key = (int(bx), int(by))

            if key not in self.cache:
                row = self.block_rows.loc[key]
                tiles = self.data.tiles(row)

                x = torch.stack([
                    to_tensor(tile) for tile in tiles
                ]).to(DEV)

                self.cache[key] = (
                    self.m.tile_features(x).detach()
                )

            tf_batch.append(self.cache[key])

        # [B, 4, 512, 32, 32]
        tf = torch.stack(tf_batch)

        # 드론 영상을 북쪽 방향으로 정렬한 뒤 중앙 176x176 사용
        u = torch.stack([
            to_tensor(north_up(crop, float(heading)))
            for crop, heading in zip(crops, headings)
        ]).to(DEV)

        pos, conf = self.m.predict(u, tf)

        return pos.cpu().numpy()


corr_loc_sf = CorrNavLoc(
    corr_model,
    b1_loc_sf.data,
    b1_loc_sf.block_rows
)

# 3. 첫 번째 SF 시작점으로 단일 예측
start = sf_starts[0]
heading = 90.0

bx = int(np.clip(np.round((start[0] - 256) / 256), 0, 14))
by = int(np.clip(np.round((start[1] - 256) / 256), 0, 14))

crop, fx, fy = sf_frames(
    float(start[0]),
    float(start[1]),
    heading
)

pred = corr_loc_sf(
    [crop],
    [heading],
    [(bx, by)]
)

print("\n===== CorrMatchNet 단일 예측 결과 =====")
print("시작점:", start)
print("선택 블록:", (bx, by))
print("Crop 크기:", np.asarray(crop).shape)
print("실제 프레임 중심:", (fx, fy))
print("예측 출력 shape:", pred.shape)
print("예측 상대 좌표:", pred[0])
print("예측값 유한 여부:", np.isfinite(pred).all())
print("타일 특징 캐시:", len(corr_loc_sf.cache))
```

```python
import time
import pandas as pd
import numpy as np

# 동일 조건으로 CorrMatchNet 내비게이션 실행
corr_sf_log = []

t0 = time.time()

corr_sf_success, corr_sf_reason, corr_sf_med_errors = nav_run(
    loc=corr_loc_sf,
    frames=sf_frames,
    starts=sf_starts,
    goals=sf_goals,
    decode_px=128,
    oracle_block=False,
    step_m=15.,
    arrive_m=20.,
    n_block=15,
    log=corr_sf_log,
    odo=ODO["vio"],
    seed=7,
    reject_inl=0,
    gate_m=0.,
    recover=None
)

elapsed = time.time() - t0

# 걸음별 결과 저장
corr_sf_steps = pd.DataFrame(corr_sf_log)
steps_path = "/kaggle/working/SF_CorrMatchNet_3D_VIO_seed7_steps.csv"
corr_sf_steps.to_csv(steps_path, index=False)

# 궤적별 결과 저장
corr_sf_traj = (
    corr_sf_steps.groupby("traj")
    .agg(
        steps=("err_m", "count"),
        median_error_m=("err_m", "median"),
        mean_error_m=("err_m", "mean"),
    )
    .reset_index()
)

corr_sf_traj["success"] = corr_sf_traj["traj"].map(
    lambda i: bool(corr_sf_success[int(i)])
)
corr_sf_traj["reason"] = corr_sf_traj["traj"].map(
    lambda i: str(corr_sf_reason[int(i)])
)

traj_path = "/kaggle/working/SF_CorrMatchNet_3D_VIO_seed7_traj.csv"
corr_sf_traj.to_csv(traj_path, index=False)

# 결과 요약
print("===== SF CorrMatchNet 3D + VIO (32 trajectories) =====")
print(f"성공: {np.sum(corr_sf_success)}/32 "
      f"({np.mean(corr_sf_success)*100:.2f}%)")
print("\n종료 이유:")
print(pd.Series(corr_sf_reason).value_counts().to_string())

print(f"\n평균 궤적별 중앙 위치오차: "
      f"{np.mean(corr_sf_med_errors):.2f} m")
print(f"총 실행 시간: {elapsed:.1f}초")
print(f"기록된 걸음 수: {len(corr_sf_steps)}")

print("\n저장 완료:")
print(steps_path)
print(traj_path)
```

```python
import pandas as pd
import numpy as np

# 각 궤적의 중앙 위치오차 계산
b1_err = (
    pd.DataFrame(sf_b1_log)
    .groupby("traj")["err_m"]
    .median()
    .rename("B1_error_m")
)

corr_err = (
    pd.DataFrame(corr_sf_log)
    .groupby("traj")["err_m"]
    .median()
    .rename("Corr_error_m")
)

paired = pd.concat([b1_err, corr_err], axis=1)

paired["B1_success"] = [
    bool(sf_b1_success[i]) for i in paired.index
]
paired["Corr_success"] = [
    bool(corr_sf_success[i]) for i in paired.index
]

paired["error_diff_Corr_minus_B1"] = (
    paired["Corr_error_m"] - paired["B1_error_m"]
)

paired["outcome"] = np.select(
    [
        paired["B1_success"] & paired["Corr_success"],
        paired["B1_success"] & ~paired["Corr_success"],
        ~paired["B1_success"] & paired["Corr_success"],
    ],
    [
        "둘 다 성공",
        "B1만 성공",
        "CorrMatchNet만 성공",
    ],
    default="둘 다 실패"
)

print("===== 궤적별 성공 결과 비교 =====")
print(paired["outcome"].value_counts().to_string())

print("\n===== 위치오차 개선 정도 =====")
print("CorrMatchNet 오차가 더 낮은 궤적:",
      (paired["error_diff_Corr_minus_B1"] < 0).sum())
print("B1 오차가 더 낮은 궤적:",
      (paired["error_diff_Corr_minus_B1"] > 0).sum())
print("동일:",
      (paired["error_diff_Corr_minus_B1"] == 0).sum())

print("\n===== 전체 궤적별 비교표 =====")
display(paired.round(2))

save_path = "/kaggle/working/SF_B1_vs_CorrMatchNet_paired.csv"
paired.to_csv(save_path, index=True)
print("\n저장 완료:", save_path)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import binomtest, wilcoxon

# 성공 여부
b1 = np.asarray(sg_success, dtype=bool)
corr = np.asarray(corr_sg_success, dtype=bool)

assert len(b1) == len(corr) == 32

# 성공/실패 조합
both_success = int(np.sum(b1 & corr))
b1_only = int(np.sum(b1 & ~corr))
corr_only = int(np.sum(~b1 & corr))
both_fail = int(np.sum(~b1 & ~corr))

print("===== Singapore 성공 여부 Paired 비교 =====")
print(f"B1만 성공: {b1_only}")
print(f"CorrMatchNet만 성공: {corr_only}")
print(f"둘 다 성공: {both_success}")
print(f"둘 다 실패: {both_fail}")

# McNemar exact test
discordant = b1_only + corr_only

if discordant > 0:
    p_value = binomtest(
        min(b1_only, corr_only),
        n=discordant,
        p=0.5,
        alternative="two-sided"
    ).pvalue
else:
    p_value = 1.0

print(f"\nMcNemar exact p-value: {p_value:.4f}")

# 궤적별 위치오차 비교
b1_err = np.asarray(sg_errors, dtype=float)
corr_err = np.asarray(corr_sg_med_errors, dtype=float)

print("\n===== 위치오차 비교 =====")
print(f"B1 평균 궤적별 중앙오차: {np.nanmean(b1_err):.2f} m")
print(f"CorrMatchNet 평균 궤적별 중앙오차: {np.nanmean(corr_err):.2f} m")
print(f"CorrMatchNet 오차가 낮은 궤적: {np.sum(corr_err < b1_err)}")
print(f"B1 오차가 낮은 궤적: {np.sum(b1_err < corr_err)}")
print(f"동일: {np.sum(b1_err == corr_err)}")

# Wilcoxon paired test
valid = np.isfinite(b1_err) & np.isfinite(corr_err)
if np.sum(valid) > 0 and np.any(b1_err[valid] != corr_err[valid]):
    w = wilcoxon(b1_err[valid], corr_err[valid])
    print(f"Wilcoxon p-value: {w.pvalue:.4f}")

# 결과 저장
paired = pd.DataFrame({
    "trajectory": np.arange(32),
    "B1_success": b1,
    "CorrMatchNet_success": corr,
    "B1_reason": sg_reason,
    "CorrMatchNet_reason": corr_sg_reason,
    "B1_median_error_m": b1_err,
    "CorrMatchNet_median_error_m": corr_err
})

paired.to_csv(
    "/kaggle/working/SG_B1_vs_CorrMatchNet_paired.csv",
    index=False
)

print("\nPaired 비교 CSV 저장 완료")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import binomtest, wilcoxon

files = {
    "SF": "/kaggle/working/SF_B1_vs_CorrMatchNet_paired.csv",
    "Singapore": "/kaggle/working/SG_B1_vs_CorrMatchNet_paired.csv"
}

results = []

for city, path in files.items():
    df = pd.read_csv(path)

    b1 = df["B1_success"].astype(bool).to_numpy()
    corr = df["CorrMatchNet_success"].astype(bool).to_numpy()

    b1_only = int(np.sum(b1 & ~corr))
    corr_only = int(np.sum(~b1 & corr))

    # 성공 여부: McNemar exact test
    discordant = b1_only + corr_only
    p_mcnemar = (
        binomtest(
            min(b1_only, corr_only),
            n=discordant,
            p=0.5
        ).pvalue
        if discordant > 0 else 1.0
    )

    # 위치오차: Wilcoxon paired test
    b1_err = df["B1_median_error_m"].to_numpy(float)
    corr_err = df["CorrMatchNet_median_error_m"].to_numpy(float)

    valid = np.isfinite(b1_err) & np.isfinite(corr_err)
    x, y = b1_err[valid], corr_err[valid]

    if len(x) and np.any(x != y):
        p_wilcoxon = wilcoxon(x, y).pvalue
    else:
        p_wilcoxon = np.nan

    results.append({
        "도시": city,
        "표본 수": len(b1),
        "B1 성공률(%)": round(b1.mean() * 100, 2),
        "Corr 성공률(%)": round(corr.mean() * 100, 2),
        "B1만 성공": b1_only,
        "Corr만 성공": corr_only,
        "McNemar p": p_mcnemar,
        "B1 위치오차(m)": np.mean(x),
        "Corr 위치오차(m)": np.mean(y),
        "위치오차 감소율(%)": (1 - np.mean(y) / np.mean(x)) * 100,
        "Wilcoxon p": p_wilcoxon
    })

summary = pd.DataFrame(results)

print("===== 두 도시 통계 검정 통합 결과 =====")
display(summary.round(6))

summary.to_csv(
    "/kaggle/working/B1_vs_CorrMatchNet_final_statistics.csv",
    index=False
)

print("저장 완료: /kaggle/working/B1_vs_CorrMatchNet_final_statistics.csv")
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

files = {
    "B1": "/kaggle/working/SF_B1_3D_VIO_seed7_steps.csv",
    "CorrMatchNet": "/kaggle/working/SF_CorrMatchNet_3D_VIO_seed7_steps.csv",
}

def team_common_eval(path):
    df = pd.read_csv(path)

    # 해당 로그에서 사용한 블록 중심
    cx = 256 + 256 * df["block_x"].to_numpy()
    cy = 256 + 256 * df["block_y"].to_numpy()

    # 블록 중심 기준 정규화 좌표
    px = (df["est_x"].to_numpy() - cx) / 128
    py = (df["est_y"].to_numpy() - cy) / 128

    gx = (df["real_x"].to_numpy() - cx) / 128
    gy = (df["real_y"].to_numpy() - cy) / 128

    # 추정 없음 및 블록 밖 실패
    no_estimate = (
        df["unknown"].to_numpy().astype(bool)
        | ~np.isfinite(px)
        | ~np.isfinite(py)
    )

    fail = (
        no_estimate
        | (np.abs(px) > 1.1)
        | (np.abs(py) > 1.1)
    )

    # 추정이 없는 경우 블록 중심 (0, 0)을 사용
    px = np.nan_to_num(px, nan=0.0, posinf=0.0, neginf=0.0)
    py = np.nan_to_num(py, nan=0.0, posinf=0.0, neginf=0.0)

    px = np.where(no_estimate, 0.0, px)
    py = np.where(no_estimate, 0.0, py)

    # 팀 공통 클리핑
    error_m = np.hypot(
        np.clip(px, -1, 1) - gx,
        np.clip(py, -1, 1) - gy
    ) * 32

    success = (~fail) & (error_m <= 10)

    result = df[["traj", "step"]].copy()
    result["team_error_m"] = error_m
    result["team_fail"] = fail
    result["team_success"] = success

    return result


sf_results = {}

for model, path in files.items():
    result = team_common_eval(path)
    sf_results[model] = result

    print(f"\n===== SF {model} =====")
    print("평가 step 수:", len(result))
    print("실패 step 수:", int(result["team_fail"].sum()))
    print("10m 성공 step 수:", int(result["team_success"].sum()))
    print("10m 성공률:", f"{result['team_success'].mean()*100:.2f}%")
    print("평균 위치 오차:", f"{result['team_error_m'].mean():.2f}m")
    print("중앙 위치 오차:", f"{result['team_error_m'].median():.2f}m")
```

```python
import numpy as np
import pandas as pd

# Singapore 로그 불러오기
sg_b1 = pd.read_csv("/kaggle/working/B1_3d_37bc_nav_log.csv")
sg_corr = pd.read_csv("/kaggle/working/SG_CorrMatchNet_3D_VIO_seed7_steps.csv")

def evaluate_steps(df, model_name):
    # 블록 중심 좌표 (512×512 블록)
    cx = 256 + 256 * df["block_x"].to_numpy(dtype=float)
    cy = 256 + 256 * df["block_y"].to_numpy(dtype=float)

    # 블록 중심 기준 정규화 좌표
    px = (df["est_x"].to_numpy(dtype=float) - cx) / 128
    py = (df["est_y"].to_numpy(dtype=float) - cy) / 128
    gx = (df["real_x"].to_numpy(dtype=float) - cx) / 128
    gy = (df["real_y"].to_numpy(dtype=float) - cy) / 128

    # unknown 또는 유효하지 않은 추정값은 추정 실패로 처리
    unknown = df["unknown"].fillna(False).astype(bool).to_numpy()
    no_estimate = unknown | ~np.isfinite(px) | ~np.isfinite(py)

    # metrics.py와 동일하게 실패 여부는 clipping 전 좌표로 판정
    fail = no_estimate | (np.abs(px) > 1.1) | (np.abs(py) > 1.1)

    # 추정 실패는 블록 중심(0, 0)으로 처리
    px = np.where(no_estimate, 0.0, px)
    py = np.where(no_estimate, 0.0, py)

    # 팀 공통 평가: 예측 좌표만 [-1, 1]로 clipping
    error_m = np.hypot(
        np.clip(px, -1, 1) - gx,
        np.clip(py, -1, 1) - gy
    ) * 32

    success_10m = (~fail) & (error_m <= 10)

    print(f"\n===== Singapore {model_name} =====")
    print(f"평가 step 수: {len(df)}")
    print(f"실패 step 수: {fail.sum()}")
    print(f"10m 성공 step 수: {success_10m.sum()}")
    print(f"10m 성공률: {success_10m.mean() * 100:.2f}%")
    print(f"평균 위치 오차: {error_m.mean():.2f}m")
    print(f"중앙 위치 오차: {np.median(error_m):.2f}m")

    return {
        "model": model_name,
        "n": len(df),
        "fail": int(fail.sum()),
        "success_10m": int(success_10m.sum()),
        "success_rate": success_10m.mean() * 100,
        "mean_error_m": error_m.mean(),
        "median_error_m": np.median(error_m),
    }

sg_b1_result = evaluate_steps(sg_b1, "B1")
sg_corr_result = evaluate_steps(sg_corr, "CorrMatchNet")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import binomtest

files = {
    "SF": {
        "B1": "/kaggle/working/SF_B1_3D_VIO_seed7_steps.csv",
        "CorrMatchNet": "/kaggle/working/SF_CorrMatchNet_3D_VIO_seed7_steps.csv",
    },
    "Singapore": {
        "B1": "/kaggle/working/B1_3d_37bc_nav_log.csv",
        "CorrMatchNet": "/kaggle/working/SG_CorrMatchNet_3D_VIO_seed7_steps.csv",
    },
}

def make_success_table(path, model):
    df = pd.read_csv(path)

    cx = 256 + 256 * df["block_x"].to_numpy(dtype=float)
    cy = 256 + 256 * df["block_y"].to_numpy(dtype=float)

    px = (df["est_x"].to_numpy(dtype=float) - cx) / 128
    py = (df["est_y"].to_numpy(dtype=float) - cy) / 128
    gx = (df["real_x"].to_numpy(dtype=float) - cx) / 128
    gy = (df["real_y"].to_numpy(dtype=float) - cy) / 128

    unknown = (
        df["unknown"].astype(str).str.lower()
        .isin(["true", "1", "yes"]).to_numpy()
    )

    no_estimate = unknown | ~np.isfinite(px) | ~np.isfinite(py)
    fail = no_estimate | (np.abs(px) > 1.1) | (np.abs(py) > 1.1)

    px = np.where(no_estimate, 0.0, px)
    py = np.where(no_estimate, 0.0, py)

    error_m = np.hypot(
        np.clip(px, -1, 1) - gx,
        np.clip(py, -1, 1) - gy
    ) * 32

    result = df[["traj", "step"]].copy()
    result["traj"] = result["traj"].astype(str)
    result["step"] = pd.to_numeric(result["step"]).astype(int)
    result[f"{model}_success"] = (~fail) & (error_m <= 10)

    return result

all_pairs = []

for city, models in files.items():
    b1 = make_success_table(models["B1"], "B1")
    corr = make_success_table(models["CorrMatchNet"], "Corr")

    paired = b1.merge(corr, on=["traj", "step"], how="inner")
    paired["city"] = city

    b1_only = int(
        (paired["B1_success"] & ~paired["Corr_success"]).sum()
    )
    corr_only = int(
        (~paired["B1_success"] & paired["Corr_success"]).sum()
    )
    both = int(
        (paired["B1_success"] & paired["Corr_success"]).sum()
    )
    neither = int(
        (~paired["B1_success"] & ~paired["Corr_success"]).sum()
    )

    discordant = b1_only + corr_only
    p_value = (
        binomtest(min(b1_only, corr_only), discordant, 0.5).pvalue
        if discordant > 0 else 1.0
    )

    print(f"\n===== {city}: paired McNemar =====")
    print(f"공통 평가 step: {len(paired)}")
    print(f"B1만 성공: {b1_only}")
    print(f"CorrMatchNet만 성공: {corr_only}")
    print(f"둘 다 성공: {both}")
    print(f"둘 다 실패: {neither}")
    print(f"McNemar exact p-value: {p_value:.6f}")

    all_pairs.append(paired)

# 두 도시 통합
combined = pd.concat(all_pairs, ignore_index=True)

b1_only = int(
    (combined["B1_success"] & ~combined["Corr_success"]).sum()
)
corr_only = int(
    (~combined["B1_success"] & combined["Corr_success"]).sum()
)
both = int(
    (combined["B1_success"] & combined["Corr_success"]).sum()
)
neither = int(
    (~combined["B1_success"] & ~combined["Corr_success"]).sum()
)

discordant = b1_only + corr_only
p_value = (
    binomtest(min(b1_only, corr_only), discordant, 0.5).pvalue
    if discordant > 0 else 1.0
)

print("\n===== SF + Singapore 통합 McNemar =====")
print(f"공통 평가 step: {len(combined)}")
print(f"B1만 성공: {b1_only}")
print(f"CorrMatchNet만 성공: {corr_only}")
print(f"둘 다 성공: {both}")
print(f"둘 다 실패: {neither}")
print(f"McNemar exact p-value: {p_value:.6f}")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

# SF와 Singapore의 궤적별 성공률 계산
traj_results = (
    pd.concat(all_pairs, ignore_index=True)
    .groupby(["city", "traj"])
    .agg(
        common_steps=("step", "count"),
        B1_rate=("B1_success", "mean"),
        Corr_rate=("Corr_success", "mean"),
    )
    .reset_index()
)

traj_results["difference"] = (
    traj_results["Corr_rate"] - traj_results["B1_rate"]
)

def trajectory_test(df, name):
    diff = df["difference"].to_numpy()
    nonzero = diff[diff != 0]

    improved = int((diff > 0).sum())
    worse = int((diff < 0).sum())
    tied = int((diff == 0).sum())

    if len(nonzero) > 0:
        stat, p = wilcoxon(
            nonzero,
            alternative="two-sided",
            method="auto"
        )
    else:
        stat, p = 0.0, 1.0

    print(f"\n===== {name}: 궤적 단위 비교 =====")
    print(f"궤적 수: {len(df)}")
    print(f"CorrMatchNet 개선 궤적: {improved}")
    print(f"B1 개선 궤적: {worse}")
    print(f"동일한 궤적: {tied}")
    print(f"B1 평균 궤적 성공률: {df['B1_rate'].mean() * 100:.2f}%")
    print(f"Corr 평균 궤적 성공률: {df['Corr_rate'].mean() * 100:.2f}%")
    print(f"평균 성공률 차이: {diff.mean() * 100:+.2f}%p")
    print(f"Wilcoxon statistic: {stat}")
    print(f"Wilcoxon p-value: {p:.6f}")

trajectory_test(
    traj_results[traj_results["city"] == "SF"],
    "SF"
)

trajectory_test(
    traj_results[traj_results["city"] == "Singapore"],
    "Singapore"
)

trajectory_test(traj_results, "SF + Singapore 통합")

# 궤적별 상세 결과 확인
display(
    traj_results.sort_values(["city", "traj"]).round(4)
)
```

```python
import numpy as np, cv2

df = b1_loc_sg.block_rows
print(df[["x_norm","y_norm","x_uccs","y_uccs","px","py","theta","x_cosa","y_sina"]].describe().T[["min","max","mean","std"]])

rng = np.random.default_rng(0)
keys = [df.index[i] for i in rng.choice(len(df), 100, replace=False)]
P, D = [], []
for k in keys:
    r = df.loc[k]
    im = cv2.cvtColor(cv2.imread(str(b1_loc_sg.data.base / r["target_path"])), cv2.COLOR_BGR2RGB)
    P.append(b1_loc_sg([im], [90.0], [k])[0]); D.append(b1_loc_sg.last_direction[0].copy())
P, D = np.array(P), np.array(D)
sub = df.loc[keys]

def report(name, cols, pred):
    G = sub[cols].to_numpy(float)
    e = np.linalg.norm(pred - G, axis=1).mean()
    base = np.linalg.norm(G - G.mean(0), axis=1).mean()
    cx = np.corrcoef(pred[:,0], G[:,0])[0,1]; cy = np.corrcoef(pred[:,1], G[:,1])[0,1]
    print(f"{name:12s} 오차 {e:.3f} | 평균기준선 {base:.3f} | corr {cx:.2f}, {cy:.2f}")

print("\n--- 위치 출력 vs 후보 정답 ---")
report("x_norm", ["x_norm","y_norm"], P)
report("uccs", ["x_uccs","y_uccs"], P)
report("px,py", ["px","py"], P)
print("\n--- 방향 출력 vs 후보 정답 ---")
report("cosa,sina", ["x_cosa","y_sina"], D)
```

```python
import numpy as np
th = np.deg2rad(sub["theta"].to_numpy(float))
ang = np.arctan2(D[:,1], D[:,0])
diff = np.rad2deg(np.angle(np.exp(1j*(ang - th))))
print("각도 차 평균(원형):", np.rad2deg(np.angle(np.mean(np.exp(1j*np.deg2rad(diff))))))
print("각도 차 집중도 R (1에 가까우면 일정 오프셋):", np.abs(np.mean(np.exp(1j*np.deg2rad(diff)))))
```
